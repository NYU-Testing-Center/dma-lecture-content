## Definitions

- entity - some thing that we store data about (the sale of a property, an incident of a dog bite, a movie)
- relation - a two-dimensional table consisting of columns and rows, with only one value at the intersection of a column and row
- composite key - two or more columns in a table that uniquely identify a row in that table; no part of a composite key may be null
- we'll also take a look at some strange things that occur when working with a poorly designed database, which includes: insertion, deletion, and update anomalies.

## Critique of the Table Structure

Is this a good way to represent the student, course, and enrollment data?

What might be a good primary key for this table? Would netid work? netid and course\_num? Or netid, course\_num, semester, and year

Can we? ...

- Show all students enrolled in a particular course? Yes.
- Show all courses in fall 2019 that have students enrolled in them? Yes

But ...

Can we show all courses in fall 2019? Not that easily ...

- This table mixes up courses, students, and enrollments which are separate entities that should each be stored in their own tables.
- Courses don't exist until someone enrolls in them, which is not so great before enrollment starts.
  - if we inserted a course before an enrollment, other columns, like netid, which is part of the composite primary key, would be NULL
  - that would be an insertion anomaly

## Insertion Anomalies

An insertion anomaly is an issue that occurs when data cannot be inserted because of an incomplete primary key (no part of a primary key can be missing because then you can't look up records by the key).

So ...

- we can't add a student to our database unless they enroll in a course
- we can't add a course unless one student enrolls in it

This means that inserting data about one entity requires inserting data about another even if that latter information is not needed.

## Another Question

Say a student changes their name, which happens frequently. Is it possible to change the name of a student?

- you'll have to change it in multiple places
- if care isn't taken, only a subset of the duplicate data may change causing inconsistent data!
- This is an update anomaly

## Update Anomalies

An update anomaly occurs when a subset of duplicate data is updated leading to inconsistent data (that is, when all instances of the same data are not changed at the same time).

- this is mainly due to redundant data
- of course, redundant data also means unnecessary usage of space

## Why Take the Time to Design Your Database?

Lack of planning and design will lead to:

- redundant (and as consequence, potentially inconsistent) data
- difficulty adding data (insertion anomalies)
- difficulty removing data (deletion anomalies)
- data prone to inconsistencies (update anomalies)

Good database design also allows for:

- the enforcing of constraints and referential integrity (given the mechanisms available in a relational database)
- growth / expandability
- overall ease of conceptual understanding

The code below shows how we can get back our original dataframe by merging together the three dataframes we split it into.

The order of the columns and rows is different, but the information is the same. In the relational model, we will see that these two dataframes would be considered the same because the row and column order isn't important or guaranteed.

```python
import pandas as pd
students_and_courses=pd.read_csv("students_and_courses.csv")
print("Original dataframe:")
print(students_and_courses)
students = students_and_courses[['netid', 'first', 'last']].drop_duplicates(subset=['netid'])
courses = students_and_courses[['course_num', 'name', 'room', 'semester', 'year']].drop_duplicates(subset=['course_num','semester','year'])
enrollments = students_and_courses[['course_num','netid','semester','year']].drop_duplicates()
originalDF=students.merge(enrollments, on='netid')
originalDF=originalDF.merge(courses,on=['course_num','semester','year'])
print("Reconstructed Dataframe:")
print(originalDF)
```
