## Review of *args

- we used \*args in a function definition to specify that a function can take an arbitrary number of arguments represented as a tuple
- \*args goes in the parameter list
- args goes in the function body to represent the tuple of arguments

Following is an example that takes a sequence of arbitrary length and returns a tuple containing the maximum and minimum values in that sequence:

```python
def my_func(*args):
  return (max(args),min(args))

print(my_func(1,2,3))
print(my_func(1,2,3,4,-2))
```

## General use of *args

The \*args syntax can be used more generally. The character \*  will unpack a tuple so it can be used in a function call, for example:

```python
t=10,20,2
for i in range(*t):
    print(i)
```

Will the following work?

```python
t=10,20,2
try:
    for i in range(t):
        print(i)
except TypeError as e:
    print("TypeError",e)
    
```

What is the difference between the following two function calls?

```python
def f(*args):
  print(args) # prints the tuple with the parentheses

def g(*args): 
  print(*args) # the tuple is unpacked

f(1,2)
g(1,2)
```

## Functions are First Class Objects

In Python, functions are first-class objects, which means they can be passed as arguments, returned as results from other functions, stored in data structures, etc. Decorators rely on this fact.

Here's an example where we pass a function name as a parameter to another function and the function passed in is then executed.

```python
def double(n):
  return n*2
def square(n):
  return n*n
def do_something(func,n):
  return func(n)
print(do_something(double,3))
print(do_something(square,3))
print(type(double))
```

## eval

You can use "eval" to convert a string into something that can be evaluated. But this is dangerous to use with user input because it sometimes can be used to run arbitrary operating system commands.

Here's an example:

```python
def square(x): return x**2
def double(x): return 2*x
s2=eval("square")
print(s2(5))
print(square(5))
print(eval("double(14)"))
print(double(14))
```

## Function Names Stored in a Dictionary

Here's an example where we store function names as values in a dictionary and then call the functions by referencing the values.

```python
def add(*args):
  r=0
  for x in args:
    r+=x
  return r

def product(*args):
  r=1
  for x in args:
    r*=x
  return r

print(add(1,2,3,4,5))
print(product(1,2,3,4,5))
d={"add":add, "product":product} # keys are strings, values are functions
print(d["add"](1,2,3,4,5))
print(d["product"](1,2,3,4,5))
```

## Inner Functions

Python  supports "inner functions" which are local to the function in which they are defined. They cannot be called from outside that function, like a local variable cannot be referenced outside the function it is defined in.

Here is an example:

```python
def greeter():
  # these functions are local to the function
  # they are wrapped in
  def hi():
    print("hi")
  def hello():
    print("hello")
  hello()
  hi()

greeter()

# Can we do the following? no
try:
  hi()
except NameError as e:
  print("NameError",e)
```

## Running a function many times

The following function runMany returns the function new\_func as its result. The function new\_func is itself defined as an inner function in terms of the arguments to runMany and takes one argument of its own.

When runMany is called with the arguments double (which is a function) and 3, it returns a new function which doubles its argument 3 times, or octuples it (that is, makes it eight times as large). Thus we can use functions to create new functions that are based on them. This is the idea behind decorators.

```python
def double(n): return 2*n
def runMany(func,n):
  # what is x? the argument to the function returned
  def new_func(x):
    for i in range(n):
      x=func(x)
    return(x)
  return new_func

octuple=runMany(double,3)
print(type(octuple))
print(octuple(5))
```

## Decorators

In what follows, we have a function called cake which returns the word "cake".

We decorate it. The new cake contains icing.

```python
def cake(): return "cake"

print(cake()) # prints "cake"
def icing(old_func):
  def new_func():
    return old_func()+' with icing'
  return new_func

cake=icing(cake)
print(cake()) # prints "cake with icing"  
```

The following is another, equivalent way to add the decorator. It is a form of syntactic sugar.

```python
def icing(old_func):
  def new_func():
    return old_func()+' with icing'
  return new_func

@icing
def cake(): return "cake"
print(cake()) # prints "cake with icing"


#we could also decorate a different function with the same decorator
@icing
def cupcake(): return "cupcake"
print(cupcake()) # prints "cupcake with icing"
```

## Some Decorators

Here is a decorator that adds exclamation points:

```python
def shout(old_f):
  def new_f(s):
    return old_f(s)+'!!!!!'
  return new_f

# decorate the str function
str=shout(str)
print(str(12))

# decorate the len function
def len2(x): return str(len(x))
len2=shout(len2)
print(len2("antidisestablishmentarianism"))
```

## Another Decorator Example

This messes up the len function.

```python
def add10(old_f):
  def new_f(x):
    return old_f(x)+10
  return new_f

len=add10(len)
print(len("hello")) # now the length is 15 rather than 5

len=add10(len)
print(len("hello")) # now it is 25
```

## Handling Multiple Arguments

Multiple arguments with decorators need to be handled properly. The following decorator code doesn't work.

```python
def shout(old_f):
  def new_f(s):
    return old_f(s)+'!!!!!'
  return new_f

@shout
def full_name(first,last):
  return f"{first} {last}"

try:
  person=full_name('Alice','Alvarez')
  print(person)
except TypeError as e:
  print("TypeError:",e)
```

We can fix this by allowing for a variable number of arguments in the decorator.

```python
def shout(old_f):
    def new_f(*args):
        res = old_f(*args)
        return res + '!!!!!'
    return new_f

@shout
def full_name(first,last):
  return f"{first} {last}"

person=full_name('Alice','Alvarez')
print(person)
```

## Uses of Decorators

The following are some uses of decorators:

- Modify arguments going into a function, or modify the function's output
- Do some things before or after we call a function
- Some examples
  - timing a function
  - caching results to speed up a function

```python
import time
def time_ns(func):
  def wrapper(*args):
      t = time.process_time_ns()
      res = func(*args)
      print(f"The function {func.__name__} took {time.process_time_ns()-t} nanoseconds to run.")
      return res
  return wrapper

# series starts with 0,1,1,2,3,5 ...
@time_ns
def fibonnaci(n):
  l=[0,1]
  for i in range(2,n):
    l.append(l[i-1]+l[i-2])
  return l[n-1]

print(fibonnaci(6))
print(fibonnaci(20))
print(fibonnaci(100))
```

## Decorators for Caching

Decorators can be used for caching. Here the lru\_cache decorator saves up to the last maxsize values returned from calls to the function decorated. If maxsize is None, it caches all of the previous calls.

In the case of the Fibonacci series, that can save a lot of time if you are calculating many values in the series, since you only have to calculate the value for a given value of n once.

```python
from functools import lru_cache

@lru_cache(maxsize=None)
def fib(n):
    if n < 2:
        return n
    return fib(n-1) + fib(n-2)

print([fib(n) for n in range(30)])

print(fib.cache_info())
```
