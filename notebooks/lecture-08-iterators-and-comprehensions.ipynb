## Looping "Under the Hood"

In Python, if an object can give up one element at a time, you can iterate over that object.

Examples of iterable objects include strings, lists, tuples, sets, range objects, and dictionaries,

## Looping Over Strings

```python
for ch in "Oscar":
    print(ch," ",end="")
print()
```

## Looping Over Lists

```python
for ele in ["Charlie", "Brown"]:
    print(ele)
```

## Looping Over Tuples

```python
for n in (1, 2, 3):
    print(n)
```

## Looping Over Ranges

```python
for i in range(25,50,5):
    print(i)
```

## Looping Over Dictionaries

```python
# dictionaries
d = {'x':1, 'y':2}
for k, v in d.items():
    print(k, v)
```

## How Does Looping Actually Work?

A for loop works like this:

1. it calls iter() on the object it's looping over
2. we get an iterator back
3. the for loop continually calls next on the iterator object
4. until a StopIteration exception occurs

The looping "interface" consists of a couple of methods:

1. iter on a container object; it will return an object that implements next
2. next on an iterator object; it will give back an element

The container and the iterator can be the same object (which is why you can return self for body of iter)

## Some definitions:

- an iterable object is an object that's capable of giving back an iterator
- an iterator is an object that implements the next method

## Some Built-In Functions

- iter(obj) - will cause obj's iter method to be called
- next(obj) - will cause obj's next method to be called

## Can We Iterate Directly on a List? No

```python
artists = ["Matisse","Picasso","O'Keeffe", "Cassatt","Renoir"]
# can we iterate with next?
try:
  next(artists)
except TypeError as e:
  print("TypeError:",e)
```

## To Iterate, We Need to Construct an Iterator

```python
artists = ["Matisse","Picasso","O'Keeffe", "Cassatt","Renoir"]
my_iterator = iter(artists)
print(artists)
print(my_iterator)

print(next(my_iterator))
print(next(my_iterator))
print(next(my_iterator))
print(next(my_iterator))
print(next(my_iterator))

# last element causes StopIteration to be "thrown"
try:
  next(my_iterator)
except StopIteration as e:
  print("StopIteration")
```

## Constructing Our Own Iterator: A Countdown Clock

```python
class Countdown:
    def __init__(self, start):
        self.cur = start
    def __iter__(self):
        return self
    def __next__(self):
        ret = self.cur
        if self.cur > 0:
            self.cur -= 1
        else:
            raise StopIteration
        return ret

c=Countdown(5)
for i in c:
    print(i)
try:
  next(c)
except StopIteration as e:
  print("StopIteration")
```

## Using "Next" Instead for the Countdown: Same Results

```python
class Countdown:
    def __init__(self, start):
        self.cur = start
    def __iter__(self):
        return self
    def __next__(self):
        ret = self.cur
        if self.cur > 0:
            self.cur -= 1
        else:
            raise StopIteration
        return ret

c=Countdown(5)
print(next(c))
print(next(c))
print(next(c))
print(next(c))
print(next(c))
try:
  next(c)
except StopIteration as e:
  print("StopIteration")
```

## The Clock is Dangerous in a Loop

Because it loops forever ...

```python
class Clock:
    def __init__(self, initTime):
        self.currentTime = initTime
    def __iter__(self):
        return self
    def __next__(self):
        ret = self.currentTime
        if self.currentTime < 23:
            self.currentTime += 1
        elif self.currentTime==23:
            self.currentTime=0
        return ret
c=Clock(10)
i=0
for hour in c:
  print(hour," ",end="")
  i+=1
  if (i%20==0): print()
  if (i==50): break # break after 50 iterations
```

## List, Set, and Dictionary Comprehensions

The following list comprehension takes the list of artists and returns another list of the ones whose name doesn't contain an s transformed to upper case.

```python
artists = ["Matisse","Picasso","O'Keeffe", "Cassatt","Renoir"]
sFreeArtists=[name.upper() for name in artists if 's' not in name ]
print(sFreeArtists)
```

## A Set Comprehension

The following code uses a set comprehension to find the squares of a list of numbers.

```python
numbers = [1, 2, 2, 3, 3, 3]
number_set={n * n for n in numbers}
print(number_set)
print(type(number_set))
```

## A Dictionary Comprehension

The following dictionary comprehension creates a dictionary with the last index in the string of each letter in "banana". This makes use of the fact that a key's value can and will be overwritten.

```python
print(list(enumerate("banana")))
# dictionary comprehension
# note that keys can be overwritten
d={ch: i for i, ch in enumerate('banana')}
print(type(d))
print(d)
```
