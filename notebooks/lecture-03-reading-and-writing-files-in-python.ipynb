## The "open" Function

The open function has 2 arguments:

- the file\_name, which is the absolute or relative path to open
- the mode, which a single character: r, w, a
  - r: the file must exist, otherwise FileNotFoundError
  - w: if it exists, overwrite it, if it doesn't, create it
  - a: append, add to existing file

open returns: a file object

a file object has following methods:

- close: closes the file handle, no args
- read: read the file's entire contents and return it as a single string
- readlines: read the file's entire contents and return it as list of strings, one per line
- readline: read the file one line at a time, location saved

The code below reads from a file and prints out its contents.

The poem is the "The Lama" by the comic poet [Ogden Nash](https://en.wikipedia.org/wiki/Ogden_Nash).

The file handle object is f, which is returned by open. You can loop through this, which gives you each line in turn. Finally, you close the file handle.

```python
fn='llama.txt'
f = open(fn, 'r')
for line in f:
    print(line,end='')
f.close()
```

Opening a file with "with" creates a block which automatically closes the file at the end of the block. The example below uses readlines to read the file into a list of strings, which each include the newline at the end of the line.

```python
with open("llama.txt", 'r') as f:
    lines = f.readlines()
print(lines)
```

The code below reads the file into a single string including the newlines. It then uses the splitlines() method to split the result into a list of strings one for each line.

```python
whole_file=open("llama.txt","r").read()
print(whole_file)
lines=whole_file.splitlines()
print(lines)
```

## Dealing with Missing Files

You can use "try ... except" to catch a FileNotFoundError if a file is not found.

```python
try:
    with open('myfile.txt', 'r') as f:
        print(f.read())
except FileNotFoundError:
    print('file not found!')
```
