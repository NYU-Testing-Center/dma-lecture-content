## Storing a point in two dimensions

We could store a point as two separate floats:

```python
x,y=4.1,5.2
print(x,y)
```

Or as a two element list or two element tuple:

```python
p=[4.1,5.2]
print(p)
p=(4.1,5.2)
print(p)
```

Or as a dictionary:

```python
p={'x':4.1, 'y':5.2}
print(p['x'],p['y'])
```

## Named Tuples

Another way to create a point is using a [named tuple](https://docs.python.org/3/library/collections.html#collections.namedtuple) from Python's collections module which allows you to access the elements of the tuple by index or by name (the latter as an attribute). Each position in the tuple gets a name.

To define a named tuple, you call "namedtuple" with the name of the type of object and the names of the fields in the object (as a list), as follows.

Here we define a type called "Point". We can do this with either named arguments or positional arguments.

```python
from collections import namedtuple
Point=namedtuple('Point',['x','y'])

# create using named (keyword) arguments 
p=Point(y=5.2,x=4.1)
print(p.x)
print(p.y)
```

In the following, we create a Point with positional arguments and access them with indices.

```python
from collections import namedtuple
Point=namedtuple('Point',['x','y'])
q=Point(7.1,-4.3)
print(q)

# three ways to access x
print(q[0])
print(q.x)
print(q[-2])

# three ways to access y
print(q[1])
print(q.y)
print(q[-1])
```

## Unpacking Named Tuples

We can unpack the named tuple like an ordinary tuple.

```python
from collections import namedtuple
Point=namedtuple('Point',['x','y'])
q=Point(7.1,-4.3)

x, y = (1.1, 2.3) # ordinary tuple
print(x, y)
x, y = q # named tuple
print(x, y)
```

## Named Tuples are Immutable

Like ordinary tuples, named tuples are immutable.

```python
from collections import namedtuple
Point=namedtuple('Point',['x','y'])
q=Point(7.1,-4.3)
t=(1.1,3.5)
try:
  t[0]=1.2
except:
  print("Can't reassign elements in an ordinary tuple.")
try:
  q[0]=1.5
except:
  print("Can't reassign elements in a named tuple.")
l=[1.1,3.5]
l[0]=1.2 # not a problem with a list
print(l)
```

## Named Tuple Type

What are the types of these objects?

```python
from collections import namedtuple
Point=namedtuple('Point',['x','y'])
q=Point(7.1,-4.3)
print(type(Point)) # Point is a type of object
print(type(q)) # this is a instance of the type

BabyCarriage=namedtuple('Point',['x','y']) # name of the type doesn't matter
r=BabyCarriage(8.3,7.1)
print(type(r)) # the type is still named Point
print(r)
```

Named tuples can be useful, but they are limited. They are immutable, which is limiting in many cases, and they have only the methods associated with tuples, as well as the ability to address elements by name.

## Classes

Classes are much more powerful and flexible (than limited objects like named tuples).

- They are a way to define a new type of objects and instantiate them
- They can define methods, which are class-specific functions
- They can represent arbitrary data structures
- Like in other object-oriented languages such as Java, they support inheritance, encapsulation, and polymorphism

## Python Programmer-Defined Objects and Classes

#### Terminology

- A class defines a type of object
- An object is an instance of a class
- When defining a class, we define attributes and methods of that class
- An attribute is a variable associated with a class and its objects
- A method is a function associated with a class and its objects

#### We can define our own classes and then create objects using them.

## Syntax of Programmer-Defined Classes

- "Class" keyword defines a class giving the class name
- Methods are defined as functions indented under the class
- "Self" refers to the object itself
- All methods (except for static methods) refer to self and optionally some other arguments
- Special "dunder" methods begin and end with double underscore

## Class Methods

Methods that start and end with double underscores have special meaning in Python. They may specify the "constructor" function, provide a mechanism for operator "overloading" (like using +, \*, etc.), and carry out string conversion.

- The \_\_init\_\_ method is used to create an object
- The \_\_str\_\_ method is used to tell the print function how to print an object
- The \_\_repr\_\_ method is used to create a string representation for use in debugging (for example), for instance when evaluating in the interactive Python shell.
- The \_\_eq\_\_ tells the == operator when two objects are equal
- Methods are called with dot notation
- Objects are created with the class name and the arguments to \_\_init\_\_

There are [many other magic or dunder methods](https://rszalski.github.io/magicmethods/) like \_\_mul\_\_  and \_\_or\_\_ .

## A Dog Class

The following is a class for dogs, that stores four attributes per dog. The age of a dog can't exceed 29.

```python
# define a class
class Dog:
  def __init__(self,name,weight,age,got_rabies_shot):
    self.name=name
    self.weight=weight
    if age<=29: # validate
      self.age=age
    else:
      raise ValueError("That is too old.")
    self.got_rabies_shot=got_rabies_shot
  def __str__(self):
    s=self.name+" is a dog that weighs "
    s+=str(self.weight)+" pounds and is "
    s+=str(self.age)+" years old;"
    if self.got_rabies_shot:
      s+=" vaccinated for Rabies."
    else:
      s+=" not vaccinated for Rabies."
    return s

# create an instance
dog1=Dog("Fido",25,1,False)
# print out the attributes
print(dog1.name)
print(dog1.weight)
print(dog1.age)
print(dog1.got_rabies_shot)

# try to create a dog that is too old
try:
  d=Dog("Spike",50,40,True)
except ValueError as e:
  print(e)
try:
  print(d)
except NameError as e:
  print(e)
```

## A Point Class

The following is a point class with two methods, ManhattanDistance and EuclideanDistance.

```python
import math
class Point: # class definition, starts with keyword "class"
    '''Creates a point on a coordinate plane with values x and y.'''

    def __init__(self, x, y): # method called when an object is created
        '''Defines x and y variables'''
        self.x = x
        self.y = y

    def __str__(self): # method called when an object is printed
        return "Point(%s,%s)"%(self.x, self.y)
    def ManhattanDistance(self,other):
        return abs(self.x-other.x)+abs(self.y-other.y)
    import math
    def EuclideanDistance(self,other):
        return math.sqrt((self.x-other.x)**2+(self.y-other.y)**2)

# create two points
p=Point(0,0)
q=Point(3,4)
print(p.x) # 0
print(p.y) # 0
print(p)
print(q)
print(p.ManhattanDistance(q))
print(p.EuclideanDistance(q))
```

## A Poker Simulation Using Python Classes

Following is a class that deals a random card from a deck and a class that deals a hand of cards from that deck. We then estimate the frequency of five-card poker flushes by simulating a lot of draws from the deck. The actual frequency considering all hands is 0.1965% which is a about two flushes per thousand hands.

```python
import random

class RandomCard:
    '''
    Class that creates objects that are random cards drawn from a deck
    with replacement
    '''
    def __init__(self):
        '''draw a card by pulling a random suit and value in that suit'''
        suits = ("Hearts","Diamonds","Spades","Clubs")
        values = ("Ace","2","3","4","5","6","7","8","9","10","Jack","Queen","King")
        self.suit = random.choice(suits)
        self.value = random.choice(values)

    def __str__(self):
        '''printing method'''
        return self.value+" of "+self.suit

    def same_suit(self,other):
        '''checks if two cards have the same suit'''
        return self.suit == other.suit

    def __eq__(self,other):
        '''checks if two cards are the same'''
        return (self.suit==other.suit) and (self.value==other.value)

class Hand:
    '''deal a hand of n cards'''
    def __init__(self,n):
        '''pull n cards at random'''
        self.cards = []
        i = 1
        while i <= n:
            card = RandomCard()
            # make sure you don't add the same card twice
            if card not in self.cards:
                self.cards.append(card)
                i += 1

    def __str__(self):
        '''print a hand; relies on str method from random_card class'''
        s = "("
        for i in range(len(self.cards)):
            s += self.cards[i].__str__()
            if i < len(self.cards)-1:
                s += ', '
            else:
                s += ")"
        return s
    def flush(self):
        check = True
        for card in self.cards[1:]:
            if not(RandomCard.same_suit(self.cards[0],card)):
                check = False
        return check

print(Hand(5)) # print a sample hand

# count the number of flushes in 100,000 hands
num_flush = 0
for i in range(100000):
    h = Hand(5)
    if h.flush(): num_flush += 1

print("Frequency of flushes:",num_flush / 100000)
```

## A Student Class

The following class represents a student at NYU with their netid and first and last names.

```python
class Student:
    def __init__(self, netid, first, last):
        self.netid = netid
        self.first = first
        self.last = last

    def __str__(self):
        # string representation (human readable)
        return f"{self.first} {self.last}"

    def __repr__(self):
        # string represenation of the actual object (for debugging purposes)
        return f"Student(netid={self.netid},first={self.first},last={self.last})"

    def __eq__(self, other):
        return self.netid == other.netid

# __init__ called
s = Student('cb2312', 'Charlie', 'Brown')
# __str__ called
print(s)
# __repr__ called
print(repr(s))
# defining __eq__ such that same netids mean same student:
clone = Student('cb2312', 'Charles', 'Brown')
print(s == clone) # True
```

## A Fraction Class

The following class represents a fraction of two integers.

It includes a method called gcf which is a static method, which means it is called on the class rather than an instance of the class.

It also implements \_\_add\_\_ and \_\_getitem\_\_ magic methods. The \_\_add\_\_ method overloads the "+" operator and \_\_getitem\_\_ implements indexing.

```python
class Fraction:

    def __init__(self, n, d):
        self.n = n
        self.d = d

    # this means that this method can be called without an instance
    # and consequently, no self is needed
    # instead, you call it using the actual class name
    # Fraction.gcf(a,b)
    # note the use of a decorator to create this
    @staticmethod
    def gcf(a, b):
        # go through every possible factor
        # check if it divides evenly into both
        # return the largest one
        cur_gcf = 1
        for factor in range(1, min(a,b)+1):
            if a % factor == 0 and b % factor == 0:
                cur_gcf = factor
        return cur_gcf

    def reduce(self):
        gcf = Fraction.gcf(self.n, self.d)
        return Fraction(self.n // gcf, self.d // gcf)

    def __str__(self):
        return "{}/{}".format(self.n, self.d)

    def __repr__(self):
        # we can call methods that are already defined
        return self.__str__()

    def add(self, other):
        new_n = (self.n * other.d) + (other.n * self.d)
        new_d = self.d * other.d
        return Fraction(new_n, new_d)

    def __add__(self, other):
        return self.add(other)

    # allow indexing! Indexing with 0 gives back the numerator
    # while indexing with 1 gives back the denominator...
    # any other index will result in an IndexError
    def __getitem__(self, i):
        if i == 0:
            return self.n
        elif i == 1:
            return self.d
        else:
            raise IndexError('Index must be 0 or 1')
    def __eq__(self,other):
       return (self.reduce().n==other.reduce().n) and (self.reduce().d==other.reduce().d)

# create and print three fractions
a = Fraction(1, 2)
b = Fraction(6, 8)
c = Fraction(1, 3)
print(f'a:{a}, b:{b}, c:{c}')

# you can index into them:
print(b[0], b[1])
# or use the attributes
print(b.n, b.d)

try:
    print(a[987]) # must be 0 or 1
except IndexError as e:
    print(type(e), e)
print(a.add(c)) # a is 1/2 and c is 1/3 so this gives 5/6
print(a+c) # again 5/6
print(a==Fraction(3,6)) # True because reduces first
print(Fraction(4,8).reduce()) # prints 1/2

# static method, called on class rather than instance
print(Fraction.gcf(9,12)) # 3
# can call on instance but is not recommended because it 
# doesn't make sense (doesn't have to do with any particular instance)
print(a.gcf(9,12)) # 3
```
