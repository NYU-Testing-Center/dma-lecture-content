## The Data: Every Inspection, Every Violation

New York City's Department of Health inspects every restaurant at least once a year. Each problem the
inspector finds is a **violation**, and each violation adds points to the inspection's **score**.
The score becomes the letter grade you see in the restaurant's window.

- The full dataset is public on NYC Open Data:
  [DOHMH New York City Restaurant Inspection Results](https://data.cityofnewyork.us/Health/DOHMH-New-York-City-Restaurant-Inspection-Results/43nn-pn8j)
- This notebook uses a **snapshot from November 2026** of the six ZIP codes around Washington Square:
  10002, 10003, 10009, 10011, 10012 and 10014
- `nyc_restaurant_inspections.csv`: one row per violation found (about 28,000 rows)
- `nyc_violation_codes.csv`: what each violation code means
- `camis` is the restaurant's permanent ID; `dba` ("doing business as") is its name

We will use this one dataset to review all five pandas lectures.

Each inspection has one score, so we first keep one row per inspection.
**A lower score is better.** 0 to 13 points earns an A, 14 to 27 a B, and 28 or more a C.

```python
insp = df.drop_duplicates(["camis", "inspection_date"])
print(insp["score"].describe())
```

<div style="
    border-left:10px solid #0b6a70;background:#e6f5f6;

  </div>
</div>

`apply` calls a function on every value. This one turns a score into the grade it should earn.

```python
def score_to_grade(score):
    if pd.isna(score):
        return None
    if score <= 13:
        return "A"
    if score <= 27:
        return "B"
    return "C"

insp = insp.assign(expected=insp["score"].apply(score_to_grade))
print(insp[["dba", "score", "expected", "grade"]].head(8))
```

`pd.crosstab` counts each combination of two columns. Almost every graded inspection lands on the
diagonal, but a few don't. Why might that be? Look them up in the data and find out.

```python
graded = insp[insp["grade"].isin(["A", "B", "C"])]
print(pd.crosstab(graded["expected"], graded["grade"]))
```

    Do Restaurants Improve After Re-inspection?
  </div>

A restaurant that doesn't earn an A on its regular (cycle) inspection is inspected again a few weeks
later. Comparing the average scores of the two kinds of visit would mislead us: only the worse
restaurants get re-inspected. Instead, `groupby("camis")["score"].diff()` compares each visit with
**the same restaurant's** previous visit.

```python
cycle = insp[insp["program"] == "Cycle Inspection"].sort_values(["camis", "inspection_date"])
cycle["change"] = cycle.groupby("camis")["score"].diff()
print(cycle.groupby("visit")["change"].agg(["mean", "count"]).round(1))
```
