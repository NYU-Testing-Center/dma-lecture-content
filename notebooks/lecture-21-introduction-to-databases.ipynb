## Defining a Database

A database is repository or organized collection of data

That could be anything. A bunch of post-it notes on my desk. Or directories and files on my laptop

However, the term database also typically refers to how that data is accessed, manipulated or managed.

- When we think of a database, we typically think of an electronic system that abstracts away how the data is physically stored on disk
- As a database user, we don't have to worry about files; instead, we work with things like:
  - an interactive graphical representation of data
  - a query language with commands issued through an interactive shell

## Database Management Systems

Access to data in a database is mediated by a Database Management System (DBMS).

- a DBMS is software that gives users and applications the ability to define, create, query and administer a database
- it translates requests for data manipulation between the clients and the database itself
- (it sits between the application accessing the data and the database)
- data is usually stored in a DBMS specific format on the file system (though this could vary based on the DBMS… for example, some may work in memory only)

## Database v. DBMS

Despite formal definitions, the term database (or database server) is often used to refer to both

- a database (an organized collection of data) or multiple databases
- and the DBMS (software for managing / allowing access to that collection of data)️

️This implies then that a DBMS can have multiple databases (it can manage multiple collections of data)

## Features of a DBMS

A DBMS typically has the following features

- some mechanism for creating a database and providing structure to the database
- a way to create, read, update and delete (CRUD)
- data access control (some method of managing access, such as authentication and/or authorization)
- administration (performance monitoring, logging, resource management, etc.)

## The Client/Server Model

Most databases fit under the client/server model

- access to the database is through a service
- the service is provided by a server:
  - a single, centralized server
  - a cluster of servers in the same physical location (and perhaps on the same computer)
  - several distributed servers in different physical locations
- the computer requesting the provided services is called the client
  - the client and server can be on the same computer (and often are for small applications)

## Implications of the Client/Server Model

This client/server architecture implies that:

- the server must be "on" (for example, a long running process waiting for requests) in order for data to be accessed
- different clients may exist for accessing the server
  - a command line client
  - a graphical client
  - a client library used within some other application like Python

## Database Security

Database security encompasses

- the facilities provided by the database for access control and encryption
- the computer(s) that the database server runs on
- the network(s) that the database server runs on

We'll mostly consider the first item, as the last two are large topics on their own (each with the potential of being a separate course)

There are special security requirements when working within a regulated environment, such as health, finance, and education. A few examples include

- HIPAA (Health Insurance Portability and Accountability Act) legislation that governs the handling of medical records and personally identifiable information ensures privacy of medical records
- FERPA Family Educational Rights and Privacy Act: legislation that governs access to educational information

## Relational Databases

Relational databases organize data in a collection of tables (relations). You can probably describe the characteristics of a relational database because we've already worked with tabular data.

- each table has named columns… with the actual data that populates the table in separate rows
- rows are sometimes referred to as records or tuples
- each table represents a kind or type of record or tuple, with every row representing an instance of that kind of record

## Relations, Rows, and Keys

Each table row has a primary key that:

- uniquely identifies that row
- rows in one table can be related to rows in other tables by:
  - adding a column that contains the other table's primary key
  - this column is called a foreign key

Using this pattern, complex data relationships can be modelled using primary keys and foreign keys.

## Properties of Relational Databases

Because relational databases are usually used to model relationships between entities:

- they are typically pretty rigid:
  - they're highly structured
  - columns and types of columns must be defined prior to inserting rows
- many relational database features deal with maintaining data integrity (such as user-defined data constraints, foreign keys, etc.)

## Transactions

A transaction is a single logical operation or unit of work in a DBMS

- this unit of work can be composed of multiple create, read, update, and delete operations (CRUD)
- for example, a single transaction may:
  - read a value from one record in one table
  - then update another record in another table using the value that was previously read

## Desirable Properties of Transactions

In order for a database to:

- be consistent / recover correctly from a crash or failure (fault tolerant)
- and maintain isolation between concurrent accesses to the database,

transactions must be: atomic, consistent, isolated, and durable (ACID)

## ACID

These properties of a transaction - atomicity, consistency, isolation and durability - are sometimes collectively referred to as ACID

- Atomicity - each transaction / (series of operations in a transaction) is all or nothing
- Consistency - every transaction ensures that the resulting database state is valid (goes from one valid state to another)
- Isolation - a failed transaction should have no effect on other transactions (even if the transactions are concurrent)
- Durability - once a transaction / operation is done, the results will remain persistent even through crash, power loss, etc.

## SQL

To query (read, update, etc.) and maintain a relational database, a domain specific language called SQL is used.

- it's a declarative programming language
  - your program describes what you want
  - rather than how you want to achieve it
  - (describe the outcome rather than the algorithm)
- SQL has been standardized by The American National Standards Institute (ANSI) and the International Standards Organization (ISO)
- despite the standardization, there are still many dialects of SQL specific to the database platform being used (and are consequently non portable)
- SQL's theoretical foundation is relational algebra (which we'll discuss later in this course)

## Declarative Programming

Remember, SQL is declarative, so it describes what you want the outcome of your program to be, not how to get to the outcome.

## Object Databases

Of course, tables aren't the only way to organize data

With the popularity of object-oriented programming in the 80's and 90's, databases began to represent information in the form of objects… and they supported features such as:

- the ability to declare custom types
- inheritance
- object versioning support (you'll see this in NoSQL document stores as well)
- relationships through pointers and the ability to retrieve all the data about an object in a single query i

In addition to object databases, hybrid object-relational databases sprang up as well.

Nowadays, object-relational mappings are the most common way to interface object-oriented languages to relational databases.

## NoSQL Databases

NoSQL databases can be categorized by how they store their data:

- key-value
- document
- graph
- triple/rdf store
- there are others (such as object, tuple store, etc.)

Note that NoSQL databases can have reliable transactions as well, but this is usually not the focus of a nosql database

## Key-Value Store

Probably the most simple conceptually… data is stored in key/value pairs. This should sound similar to some data structures that you've seen before. →

- maybe a hash
- or a dictionary
- or an associative array

They're typically good at scaling to handle large amounts of data and dealing with high volumes of changes in data

## Key-Value Store Examples

Some key value databases include:

- HBase (key value and columnar, distributed, large data sets)
- Redis (a popular backend for queuing)
- Memcache (as the name implies, typically used for caching)
- Riak
- many others

## Document Stores

As you might guess by the name, document stores store and allow searching of data semi-structured documents.

- think JSON (but there are many possible formats, such as XML, YAML, etc.)
- or… a richer key-value store (there's meta data within the document… the keys are usually meaningful)
- typically, no schema is required (that is, data types of values are inferred from values)
- typically, semi structured (documents, property names, etc… do not have to be pre-defined)
- some document stores are strong on availability and scaling (through replication/redundancy and sharding/separating large databases into smaller ones)

## Document Stores, Continued

Document stores are particularly good for applications where flexible data storage or constantly changing data storage is required.

Two of the most popular NoSQL document stores are:

- CouchDB
- MongoDB

Of course, there are many others.

Document Store Use Cases

- applications that require semi structured data / data that does not have rigid requirements (perhaps a resume)
- again, large volumes of data
- fluid data or data whose structure is prone to change

## Graph Databases

In a Graph Database:

- data is stored as nodes and edges of a graph
- edges represent relationships that directly link items in the database
- useful for modeling complex hierarchical relationships, networks, or any data set that fits into a graph-like structure

The best known graph database is Neo4j.

## "NewSQL"

Relational databases that provide the same scalability as NoSQL databases and ACID guarantees of relational databases

Examples: Google Spanner, Apache Ignite

## A Relational Database Management System vs Pandas?

Not really the same at all! An RDBMS is

- a service!
- has multi-client / concurrent access
- has authentication and authorization / role based access transactions
- has ACID compliance
- has replication, logging, clustering, etc.

However, in some cases, with relatively small groups of tables, Pandas is all you need; you can "pickle" your dataframes to persist them. Pickling is a Python function to store objects in memory in files for later use.
