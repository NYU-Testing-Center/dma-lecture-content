## Window Functions and Aggregation Functions Do Different Things

<div style="text-align:center;margin:0.4em 0;"><img src="images/DZBrPEA0K4qUQXS1xhr8oU8i.png" alt="" style="max-width:100%;width:512px;height:auto;"></div>

## Aggregation Functions

- Reduce many rows to one using an aggregate function such as sum(), count(), avg()
- Therefore typically reduce the number of rows.

## Window Functions

- Keep the same number of rows
- Calculate new columns based on groups of rows called “windows”
- Can do aggregations within windows or pull out particular values
- “Over” clause specifies window for a new variable

The query below ranks employees within Job title by descending salary. Note that it does not reduce the number of rows. It does have the side effect of sorting by job title.

```python
import sqlite3
import pandas as pd 
conn=sqlite3.connect("salaries.db")
query='''
select EmployeeName, JobTitle, TotalPay, rank()
over (partition by JobTitle order by TotalPay desc) as pay_rank
from Salaries
'''
df= pd.read_sql_query(query,conn)
print(df)
```

The query below finds the quintiles on total pay within each job title. These are five roughly equal sized groups.

```python
import sqlite3
import pandas as pd 
conn=sqlite3.connect("salaries.db")
query='''
select EmployeeName, JobTitle, TotalPay, ntile(5) 
over (partition by JobTitle order by TotalPay desc)
quintile from Salaries order by JobTitle, quintile asc
'''
df= pd.read_sql_query(query,conn)
print(df)
```

This query finds the salary difference between successive employees sorted by salary within job title.

```python
import sqlite3
import pandas as pd 
conn=sqlite3.connect("salaries.db")
query='''
select sf2.EmployeeName, JobTitle, TotalPay, next_salary, TotalPay-next_salary as salary_difference from
(select EmployeeName, JobTitle, TotalPay, lead(TotalPay,1)
over (partition by JobTitle order by TotalPay desc)
as next_salary from Salaries) as sf2
'''
df= pd.read_sql_query(query,conn)
print(df)
```

This query finds the pay rank among all employees irrespective of title The window is the.entire table, sorted.

```python
import sqlite3
import pandas as pd 
conn=sqlite3.connect("salaries.db")
query='''
select EmployeeName, JobTitle, TotalPay, rank()
over (order by TotalPay desc) as pay_rank from salaries
'''
df= pd.read_sql_query(query,conn)
print(df)
```

The query below calculates average total pay within each of the five quintiles.

```python
import sqlite3
import pandas as pd 
conn=sqlite3.connect("salaries.db")
query='''
select *, avg(TotalPay) over (partition by quintile) as quintileAvgPay
from (select EmployeeName, JobTitle, TotalPay, ntile(5)
over (order by TotalPay desc) as quintile from salaries)
'''
df= pd.read_sql_query(query,conn)
print(df)
```

The query below finds the difference between total pay and average total pay within each quintile.

```python
import sqlite3
import pandas as pd 
conn=sqlite3.connect("salaries.db")
query='''
select EmployeeName, TotalPay, AveragePay, quintile,
TotalPay-AveragePay as pay_difference
from (select *, avg(TotalPay)
over (partition by quintile) as AveragePay
from (select EmployeeName, JobTitle, TotalPay, ntile(5)
over (order by TotalPay desc) as quintile from salaries) sf2) sf
'''
df= pd.read_sql_query(query,conn)
print(df)
```

  </div>

This query finds the average, minimum, and maximum total pay by quintile.

It uses a couple of subqueries.

```python
import sqlite3
import pandas as pd 
conn=sqlite3.connect("salaries.db")
query='''
select distinct quintile, average_pay, max_pay, min_pay
from (select *, avg(TotalPay) over (partition by quintile) as average_pay,
max(TotalPay) over (partition by quintile) as max_pay,
min(TotalPay) over (partition by quintile) as min_pay
from (select EmployeeName, JobTitle, TotalPay, ntile(5)
over (order by TotalPay desc) as quintile from salaries))
order by quintile
'''
df= pd.read_sql_query(query,conn)
print(df)
```

This query does the same thing as the query in the previous slide, but it uses group by instead of window functions alone. It still uses a window function to define the quintiles, though.

```python
import sqlite3
import pandas as pd 
conn=sqlite3.connect("salaries.db")
query='''
select quintile, avg(TotalPay), max(TotalPay), min(TotalPay) from
(select EmployeeName, JobTitle, TotalPay, ntile(5)
over (order by TotalPay desc) as quintile from salaries)
group by quintile order by quintile
'''
df= pd.read_sql_query(query,conn)
print(df)
```
