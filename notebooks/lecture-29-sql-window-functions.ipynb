## Window Functions and Aggregation Functions Do Different Things

<div style="text-align:center;margin:0.4em 0;"><img src="https://raw.githubusercontent.com/NYU-Testing-Center/dma-lecture-content/main/notebooks/images/DZBrPEA0K4qUQXS1xhr8oU8i.png" alt="" style="max-width:100%;width:512px;height:auto;"></div>

## Aggregation Functions

- Reduce many rows to one using an aggregate function such as sum(), count(), avg()
- Therefore typically reduce the number of rows.

## Window Functions

- Keep the same number of rows
- Calculate new columns based on groups of rows called “windows”
- Can do aggregations within windows or pull out particular values
- “Over” clause specifies window for a new variable

The query below ranks employees within Job title by descending salary. Note that it does not reduce the number of rows. It does have the side effect of sorting by job title.

```python
import sqlite3
import pandas as pd 
conn=sqlite3.connect("salaries.db")
query='''
select EmployeeName, JobTitle, TotalPay, rank()
over (partition by JobTitle order by TotalPay desc) as pay_rank
from Salaries
'''
df= pd.read_sql_query(query,conn)
print(df)
```

The query below finds the quintiles on total pay within each job title. These are five roughly equal sized groups.

```python
import sqlite3
import pandas as pd 
conn=sqlite3.connect("salaries.db")
query='''
select EmployeeName, JobTitle, TotalPay, ntile(5) 
over (partition by JobTitle order by TotalPay desc)
quintile from Salaries order by JobTitle, quintile asc
'''
df= pd.read_sql_query(query,conn)
print(df)
```

This query finds the salary difference between successive employees sorted by salary within job title.

```python
import sqlite3
import pandas as pd 
conn=sqlite3.connect("salaries.db")
query='''
select sf2.EmployeeName, JobTitle, TotalPay, next_salary, TotalPay-next_salary as salary_difference from
(select EmployeeName, JobTitle, TotalPay, lead(TotalPay,1)
over (partition by JobTitle order by TotalPay desc)
as next_salary from Salaries) as sf2
'''
df= pd.read_sql_query(query,conn)
print(df)
```

The query below finds the difference between total pay and average total pay within each quintile.

```python
import sqlite3
import pandas as pd 
conn=sqlite3.connect("salaries.db")
query='''
select EmployeeName, TotalPay, AveragePay, quintile,
TotalPay-AveragePay as pay_difference
from (select *, avg(TotalPay)
over (partition by quintile) as AveragePay
from (select EmployeeName, JobTitle, TotalPay, ntile(5)
over (order by TotalPay desc) as quintile from salaries) sf2) sf
'''
df= pd.read_sql_query(query,conn)
print(df)
```

This query finds the average, minimum, and maximum total pay by quintile.

It uses a couple of subqueries.

```python
import sqlite3
import pandas as pd 
conn=sqlite3.connect("salaries.db")
query='''
select distinct quintile, average_pay, max_pay, min_pay
from (select *, avg(TotalPay) over (partition by quintile) as average_pay,
max(TotalPay) over (partition by quintile) as max_pay,
min(TotalPay) over (partition by quintile) as min_pay
from (select EmployeeName, JobTitle, TotalPay, ntile(5)
over (order by TotalPay desc) as quintile from salaries))
order by quintile
'''
df= pd.read_sql_query(query,conn)
print(df)
```

This query does the same thing as the query in the previous slide, but it uses group by instead of window functions alone. It still uses a window function to define the quintiles, though.

```python
import sqlite3
import pandas as pd 
conn=sqlite3.connect("salaries.db")
query='''
select quintile, avg(TotalPay), max(TotalPay), min(TotalPay) from
(select EmployeeName, JobTitle, TotalPay, ntile(5)
over (order by TotalPay desc) as quintile from salaries)
group by quintile order by quintile
'''
df= pd.read_sql_query(query,conn)
print(df)
```

This view keeps only the summer medals (bronze, silver, or gold), and only the name, year, event, and medal columns.

This view uses <code>lag()</code> to attach to each medal the athlete's previous medal in the same event, partitioning by name and event. Rows with no previous medal (NULL) are dropped.

```python
conn2.execute("drop view if exists medal_lag")
conn2.execute('''
create view medal_lag as
select * from (select *, lag(medal,1)
over (partition by name, event order by name, event, year)
as lastmedal from summer_medals_only) l
where lastmedal is not NULL order by name, event, year
''')

query='''
select * from medal_lag order by year desc
'''
df= pd.read_sql_query(query,conn2)
print(df)
```

This aggregated view has one row per month with the average daily volume. Note <code>%m</code> is the month; <code>%M</code> would be minutes.

```python
conn3.execute("drop view if exists monthly_averages")
conn3.execute('''
create view monthly_averages as
select strftime('%Y',date) as year,
strftime('%m',date) as month, avg(volume) as avg
from msft
group by year, month order by year, month
''')

query='''
select * from monthly_averages
'''
df= pd.read_sql_query(query,conn3)
print(df)
```

This view splits the date into year, month, and day and keeps the volume.

```python
conn3.execute("drop view if exists volumes_only")
conn3.execute('''
create view volumes_only as
select strftime('%Y',date) as year,
strftime('%m',date) as month,
strftime('%d',date) as day, volume from msft
''')

query='''
select * from volumes_only
'''
df= pd.read_sql_query(query,conn3)
print(df)
```

    Daily Volume vs. the Monthly Average, with a Join
  </div>
</div>

Joining the two views shows each day's volume as a percentage of its month's average.

```python
query='''
select volumes_only.*, monthly_averages.avg as monthly_average_volume,
round(volumes_only.volume*100.0/monthly_averages.avg,2) as percentage_of_average
from volumes_only inner join monthly_averages
on volumes_only.year=monthly_averages.year and volumes_only.month=monthly_averages.month
'''
df= pd.read_sql_query(query,conn3)
print(df)
```

    The Same Thing with a Window Function
  </div>

A window function partitioned by year and month gives the same result with one view and no join.

```python
query='''
select *, round(volume*100.0/monthly_average_volume,2) as percentage_of_average
from
(select *, avg(volume) over (partition by year, month) as monthly_average_volume
from volumes_only) v
'''
df= pd.read_sql_query(query,conn3)
print(df)
```

Adding <code>order by</code> inside the window makes <code>sum()</code> a running total that restarts each month.

```python
query='''
select *, sum(volume) over (partition by year, month order by year, month, day)
as running_sum
from volumes_only
'''
df= pd.read_sql_query(query,conn3)
print(df)
```

This view keeps year, month, day, and the adjusted closing price. The next query adds the maximum closing price of each month to every row.

```python
conn3.execute("drop view if exists closing_prices_only")
conn3.execute('''
create view closing_prices_only as
select strftime('%Y',date) as year,
strftime('%m',date) as month,
strftime('%d',date) as day, adj_close from msft
''')

query='''
select *, max(adj_close) over (partition by year, month) as maxprice
from closing_prices_only order by year, month, day
'''
df= pd.read_sql_query(query,conn3)
print(df)
```

A frame of <code>rows between 4 preceding and current row</code> averages today's price with the four trading days before it.

This view uses <code>lag()</code> to attach the previous day's close (<code>yclose</code>); the query then computes the percentage change from one day to the next.

```python
conn3.execute("drop view if exists one_day_price_lag")
conn3.execute('''
create view one_day_price_lag as
select *, lag(adj_close,1) over (order by year, month, day) as yclose
from closing_prices_only
''')

query='''
select *, (adj_close-yclose)*100/yclose as return from one_day_price_lag
'''
df= pd.read_sql_query(query,conn3)
print(df)
```
