You can have a particular column serve as the index by using the set\_index method. This will cause the column to no longer exist, except as the index.

```python
import pandas as pd
cols = ["Year", "State", "Title", "Employment", "Salary"]

data = [[2016, "CA", "Web Dev", 22650, 82930],
        [2016, "CA", "DB Admin", 12370, 93960],
        [2016, "NY", "Web Dev", 11410, 81140],
        [2016, "NY", "DB Admin", 6650, 91720],
        [2017, "CA", "Web Dev", 21150, 84270],
        [2017, "CA", "DB Admin", 12030, 95630],
        [2017, "NY", "Web Dev", 11900, 82360],
        [2017, "NY", "DB Admin", 7170, 94330],
        [2018, "CA", "Web Dev", 20170, 86160],
        [2018, "CA", "DB Admin", 10970, 100890],
        [2018, "NY", "Web Dev", 12030, 79880],
        [2018, "NY", "DB Admin", 7100, 99000]]

df = pd.DataFrame(data, columns=cols)
# note no index was specified so we get an automatic one
print(df)
df2=df.copy()
df2=df.set_index('Year')
print(df2)
```

The code below strips away the outermost index (for year) by specifying a particular year and creates a dataframe consisting only of the data for that year. The smaller dataframe has only one level of index.

```python
import pandas as pd
cols = ["Year", "State", "Title", "Employment", "Salary"]

data = [[2016, "CA", "Web Dev", 22650, 82930],
        [2016, "CA", "DB Admin", 12370, 93960],
        [2016, "NY", "Web Dev", 11410, 81140],
        [2016, "NY", "DB Admin", 6650, 91720],
        [2017, "CA", "Web Dev", 21150, 84270],
        [2017, "CA", "DB Admin", 12030, 95630],
        [2017, "NY", "Web Dev", 11900, 82360],
        [2017, "NY", "DB Admin", 7170, 94330],
        [2018, "CA", "Web Dev", 20170, 86160],
        [2018, "CA", "DB Admin", 10970, 100890],
        [2018, "NY", "Web Dev", 12030, 79880],
        [2018, "NY", "DB Admin", 7100, 99000]]

df = pd.DataFrame(data, columns=cols)
df=df.set_index(['Year','State'])
print(df)
df2017=df.loc[2017]
print(df2017)
```

After setting 'Year' and 'State' as indices, we use the loc method to retrieve the records for California in the year 2017.

There are two ways to drill down using both indices. In the first way, you pass "loc" a tuple, and in the second way, two arguments. We show both ways, which give the same output.

```python
import pandas as pd
cols = ["Year", "State", "Title", "Employment", "Salary"]

data = [[2016, "CA", "Web Dev", 22650, 82930],
        [2016, "CA", "DB Admin", 12370, 93960],
        [2016, "NY", "Web Dev", 11410, 81140],
        [2016, "NY", "DB Admin", 6650, 91720],
        [2017, "CA", "Web Dev", 21150, 84270],
        [2017, "CA", "DB Admin", 12030, 95630],
        [2017, "NY", "Web Dev", 11900, 82360],
        [2017, "NY", "DB Admin", 7170, 94330],
        [2018, "CA", "Web Dev", 20170, 86160],
        [2018, "CA", "DB Admin", 10970, 100890],
        [2018, "NY", "Web Dev", 12030, 79880],
        [2018, "NY", "DB Admin", 7100, 99000]]

df = pd.DataFrame(data, columns=cols)
df=df.set_index(['Year','State'])
print(df.loc[2017,'CA'])
print(df.loc[(2017,'CA')])
```

In the example below, we group the salary column by year, and then obtain the mean salary by year and the total salary by year.

```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)
cols = ["Year", "State", "Title", "Employment", "Salary"]

data = [[2016, "CA", "Web Dev", 22650, 82930],
        [2016, "CA", "DB Admin", 12370, 93960],
        [2016, "NY", "Web Dev", 11410, 81140],
        [2016, "NY", "DB Admin", 6650, 91720],
        [2017, "CA", "Web Dev", 21150, 84270],
        [2017, "CA", "DB Admin", 12030, 95630],
        [2017, "NY", "Web Dev", 11900, 82360],
        [2017, "NY", "DB Admin", 7170, 94330],
        [2018, "CA", "Web Dev", 20170, 86160],
        [2018, "CA", "DB Admin", 10970, 100890],
        [2018, "NY", "Web Dev", 12030, 79880],
        [2018, "NY", "DB Admin", 7100, 99000]]

df = pd.DataFrame(data, columns=cols)
# this groups the salaries by year and then takes their mean
grouped=df["Salary"].groupby(df['Year'])
print(type(grouped))
print(grouped.mean())
print(grouped.sum())
```

You can use the aggregate method to apply a custom aggregation using a function.

In the code below, we define our own mean function and pass it to aggregate.

```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)
cols = ["Year", "State", "Title", "Employment", "Salary"]

data = [[2016, "CA", "Web Dev", 22650, 82930],
        [2016, "CA", "DB Admin", 12370, 93960],
        [2016, "NY", "Web Dev", 11410, 81140],
        [2016, "NY", "DB Admin", 6650, 91720],
        [2017, "CA", "Web Dev", 21150, 84270],
        [2017, "CA", "DB Admin", 12030, 95630],
        [2017, "NY", "Web Dev", 11900, 82360],
        [2017, "NY", "DB Admin", 7170, 94330],
        [2018, "CA", "Web Dev", 20170, 86160],
        [2018, "CA", "DB Admin", 10970, 100890],
        [2018, "NY", "Web Dev", 12030, 79880],
        [2018, "NY", "DB Admin", 7100, 99000]]

df = pd.DataFrame(data, columns=cols)
f = lambda group: sum(group) / len(group)
print(df.groupby('State').aggregate({'Salary': f}))
```
