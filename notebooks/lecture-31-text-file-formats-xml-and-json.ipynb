## Methods of Storing Data in Text Files

People have often stored data in text files in order to post it on web sites and otherwise distribute it. There are four methods that are in common use.

- CSV (comma-separated values) files: data is separated by commas and multiple fields are stored on each line. Sometimes strings are quoted. Frequently a different delimiter is used, often a tab. (Tabs are actually better than commas because they are less likely to be in the data itself; sometimes a tab-separated file is called a TSV file.)
- Fixed-width files: These store each field in a fixed number of columns on each line. For instance, the first field could be in columns 1-10, the second in columns 11-15, the third in columns 16-20. This is one of the oldest ways of storing data in a text file. Such text files are the least user-readable of all of the formats we will discuss.
- Structured, tree-formats: these are more modern and include XML and JSON.

## Extensible Markup Language (XML)

XML is a structured, tag-based markup file format for describing and storing structured data in text files, typically large ones. It is domain-specific and the tags are defined according to the domain, hence the "extensible." Because of its use of opening and closing tags surrounding data, it looks similar to HTML, except the tags are different.

However, in recent years it has been losing market share to JSON (Javascript Object Notation). JSON looks almost identical to a printed Python dictionary or list of dictionaries. More on JSON below.

Both XML and JSON are much more human-readable than CSV files, the other main way that data is distributed in text files.

However, XML is still important since it is used in many domains. For instance, [Scalable Vector Graphics (SVG)](https://developer.mozilla.org/en-US/docs/Web/SVG/Tutorial/Basic_Shapes) files are in a dialect of XML. Modern web browsers can directly display SVG files.

However, we can use Python's own JSON reading capabilities to turn the JSON into a dictionary and then strip one level of structure off that list.

After we do that, we can read the data into pandas using pandas's DataFrame.from\_dict method.

````python
import json
import pandas as pd 
with open("books.json") as f:
  d=json.load(f)
print(d)
print(type(d))
titles=d['bookList']
print(titles)
print(type(titles))
print(pd.DataFrame.from_dict(titles))
````

What if we wanted to output the data in a fixed-width format? We can use Python f-strings and format specifiers to do that.

````python
import json
import pandas as pd 
with open("books.json") as f:
  d=json.load(f)
titles=d['bookList']
print(f"{'title':20s}{'author':20s}{'price':5s}")
for book in titles:
  print(f"{book['title']:20s}{book['author']:20s}{book['price']:5.2f}")
````

We can use the same format as in the previous slide to write out a file in fixed width format.

````python
# This example works on its own scratch copy of the data, so it runs the
# same way no matter which of the other examples you ran first.
from pathlib import Path
Path("books-scratch.txt").unlink(missing_ok=True)
import json
import pandas as pd 
with open("books.json") as f:
  d=json.load(f)
titles=d['bookList']
print(pd.DataFrame.from_dict(titles))
f=open("books-scratch.txt","w")
f.write(f"{'title':20s}{'author':20s}{'price':5s}\n")
for book in titles:
  f.write(f"{book['title']:20s}{book['author']:20s}{book['price']:5.2f}\n")
f.close()
````
