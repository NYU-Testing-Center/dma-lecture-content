## Review: Defining a Relational Database

The term relational database doesn't stem from the fact that a database contains entities and their relationships.

- instead, this terminology comes from the idea of a relation: a mathematical notion of what's essentially a table
- so… a relational database is a collection of "relations" (a bunch of tables)
- it's a little more nuanced than this, as relations have certain properties
- most of those properties are directly applicable to the tables that represent them
- this idea of a relational data model came from Edgar (E.F.) Codd in the early 1970's… and has been refined by others since…

## Relations

A relation can be defined as:

- a table with columns (called attributes) and rows (called tuples)
- an actual row of data is an instance of a relation (like an instance of an entity)
- That just seems like a regular old table or a spreadsheet… what's new?

The rows and columns must have certain properties (a DBMS will enforce these properties using constraints!)

## Columns

- columns must have unique names within a table (though, two different tables can have a column with the same name… and we can use aliasing to deal with this if both tables and their common columns are used in the same query)
- the values in a column must be from a domain (that is, columns have a type)
- when working with columns, the column position has no meaning (in practice, there is an ordering necessary when running certain queries, but this ordering does not affect the meaning of the data; the columns could be in any order an mean the same thing)

## Rows

- again, as we've described entities, the intersection of a column and row can only contain a single value.
- there are no duplicate rows in a relation (hence, primary keys)
- once again, like columns, rows can be viewed in any order (position does not affect meaning)

## Normalization

Normalization is the process of structuring a relational database to:

- reduce redundant data / use less space
- improve quality and integrity of data
- make it easier to change data (avoid insert, update and delete anomalies)
- allow for growth, evolution of data (reduce the need for restructuring tables completely when new data is introduced)
- make referential integrity constraints easier to enforce
- create clear and easy-to-understand structure that closely models the situation that the data represents

## History of Normalization Theory

This process was proposed by Edgar F Codd, in a paper called "A Relational Model of Data for Large Shared Data Banks"

- Codd added more "normal forms" in subsequent works
- and eventually collaborated with Raymond F. Boyce
- CJ Date's works also added to the interpretation of modeling data

## Normalization Controversy

As both relational databases and the process of normalization have evolved, some concepts are actively being debated

- for example, atomicity is cited in the original work, but its meaning is actually up for interpretation (how atomic?)
  - a string 'hello', can be decomposed into individual characters, but should it?
  - timestamp can be decomposed into year, month, day, hours, minutes and seconds)
- does null violate a rule of only having a set data type for a column or is it accepted as a special marker for denoting missing values

With that said, keep practicality and the goals of normalization in mind rather than a rigid adherence to rules or a philosophical viewpoint

## First Normal Form (1NF)

The first normal form specifies:

- no repeating groups
  - meaning of repeating groups has evolved since original paper / article
  - it can mean… no repeated columns or columns that contain multiple values (atomicity)
  - basically, when none of a relation's domains (types of columns) have a set as a possible value
  - every row and col intersection only contains one value from the domain (again, atomic)
- attributes are defined and…
- attributes depend on a primary key

## First Normal Form (2)

Another way to look at it is… 1NF is related to the shape of a record type:

- all rows must have same number of fields
- atomicity - a single column can't have multiple values
  - 123-456-7890, 555-555-5555

Some additional attributes of 1NF:

- ordering of columns or rows shouldn't be relevant
- no duplicate rows (primary key again!)

## Second Normal Form (2NF)

A relation is in second normal form if:

1. it's in 1NF
2. AND there are no partial dependencies
  - a partial dependency is when a non-prime attribute has a functional dependency on a part of composite primary key (or even on a candidate key)
3. if a table has a primary key that consists of only one column, it's already in 2NF
4. deals with most insertion / deletion anomalies left over from 1NF

## Superkeys

A superkey is any set of attributes that can identify a row. Trivially, the set of all attributes in a table is a superkey. Some of these attributes are most likely extraneous (are not needed to identify a row) but nevertheless can form part of a superkey

## Candidate Keys

A candidate key is a minimal superkey, that is, a superkey that doesn't contain a smaller superkey.

There may be several candidate keys in a table. You select one of the candidate keys to be the primary key; hence the name "candidate" since each candidate key is eligible to serve as the primary.

An irrelevant joke:

"If nominated, I will not run ... If elected, I will not serve."

## Functional Dependencies

A [functional dependency](https://opentextbc.ca/dbdesign01/chapter/chapter-11-functional-dependencies/) is one when one attribute's value depends on / is determined by another attribute's value.

- if you have two attributes, X and Y…
- for every valid instance of X, that value of X uniquely determines the value of Y
- Y is functionally dependent on X if every row having the same value X has the same value for Y
- or, more simply: if you know the value of X, you can determine the value of Y
- note two different values for X may produce the same Y

## Third Normal Form (3NF)

A relation is in third normal form if:

1. It's in 2NF
2. No non-prime attribute is dependent on the primary key through another non-prime attribute
3. If there is only one non-key attribute, it is already in 3NF.

For instance, one might have a table with the following columns: employee, department, and location. If the employee field is the key, and the employee field determines the department, and the location is determined by the department, then rule #2 above is violated. The information about location is duplicated each time the department appears in the table.

## The Normalization "Pledge"

2NF and 3NF deal with the relationships between keys and non-keys

- Via Bill Kent's [A Simple Guide to Five Normal Forms](https://dl.acm.org/doi/10.1145/358024.358054)
- "[Every] non-key [attribute] must provide a fact about the key, the whole key, and nothing but the key.

Does this cover 1NF, 2NF, and 3NF? Yes.

- 1NF: "the key" implies we have a key that uniquely identities each record
- 2NF: "the whole key" means we have no partial dependencies, that is, attributes that are dependent on only part of the key
- 3NF: "nothing but the key" means it does not depend on non-key attributes.

## Boyce-Codd Normal Form (BCNF)

A relation R is in BCNF (sometimes called 3.5NF) if for every non-trivial functional dependency X->B, X is a superkey of R. A superkey is a set of attributes that can uniquely identify each row in R. (Note a trivial functional dependency is of the form X->Y where Y is a subset of X.)

## Fourth and Fifth Normal Form (4NF and 5NF)

Typically, either 3NF or BCNF is good enough for preventing insertion, update, and deletion anomalies.

However, there are additional normal forms:

- 4NF: 3NF + "a record type should not contain two or more independent multi-valued facts about an entity"
  - see [pizza example on wikipedia](https://en.wikipedia.org/wiki/Fourth_normal_form)
  - usually the answer is to split into two or more tables
- [5NF](https://en.wikipedia.org/wiki/Fifth_normal_form): 4NF + "information content cannot be reconstructed from several smaller record types, i.e., from record types each having fewer fields than the original record"
