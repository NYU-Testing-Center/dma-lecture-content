## A Neat "Pythonic" Sorting Program

This program for Quicksort is borrowed from [this tutorial](https://cs231n.github.io/python-numpy-tutorial/).

I think it nicely illustrates the power of Python.

How do you think it works?

```python
def quicksort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quicksort(left) + middle + quicksort(right)

print(quicksort([3,6,8,10,1,2,1]))
```

## Assignment and Operators

Python variable names can have alphanumeric characters and the underscore; they must begin with an alphabetic character or an underscore.

The following code illustrates the assignment operator and the basic arithmetic operators including integer division and modulo.

It also illustrates f-strings in Python, which allow you to interpolate any expression into a string by enclosing it in braces. You do this by prefixing the string with the letter "f".

```python
x=20
y=9
print(f"The value of x is {x} and the value of y is {y}.")
print(f"Their sum is {x+y}.") # 29
print(f"x minus y is {x-y}.") # 11
print(f"x times y is {x*y}.") # 180
print(f"x squared is {x**2}.") # 400
print(f"x divided by y is {x/y}.") # 2.2222222222222223
print(f"x divided by y (using integer division) is {x//y}.") # 2
print(f"x modulo y is {x%y}") # 1
```

## Augmented Assignment Operators

Python also has augmented assignment which combine an operation like addition or subtraction with assignment. These are best illustrated by example.

```python
x=10
print(f"Initally, x is {x}.")
x+=5 # same as x=x+5
print(f"After adding 5 to x, x is now {x}.")
x-=2 # same as x=x-2
print(f"After subtracting 2 from x, x is now {x}.")
x*=2
print(f"After doubling 2, x is now {x}.")
x%=6 # same as x=x%6
print(f"After setting x to the remainder when dividing x by 6, x is now {x}.")
x**=3 # same as x=x**3
print(f"After cubing x, x is now {x}.")
x//=3 # same as x=x//3
print(f"After dividing x by 3 (integer division), x is now {x}.")
x/=(0.2) # same as x=x/(0.2)
print(f"After dividing x by 0.2, x is now {x}.")
```

## Indentation in Python

Python uses indentation to create code blocks, as in the following definition of a prime testing function.

The use of whitespace/indentation for code blocks makes Python code very readable and is part of the reason for Python's popularity. The downside is that it is very easy to make indentation mistakes.

```python
def isPrime(x): # indentation level 0
    # check divistors up to the square root of x
    if x<2: return False # indentation level 1
    for i in range(2,int(x**0.5)+1): # indentation level 1
        if x%i==0: # indentation level 2
            return False # indentation level 3
    return True # indentation level 1
print(isPrime(13)) # True
print(isPrime(14)) # False
```

The code below finds all of the primes less than 30.

It does it by three methods:

1. Using a loop
2. With a "list comprehension"
3. Using the filter function

We will be covering the last two methods in more detail later. This simple example indicates something of the "multi-paradigmatic" nature of Python; it supports structured, object-oriented, and functional programming.

Note that the design of Python was partially based on [work at NYU](https://www.nytimes.com/2009/03/04/science/04schwartz.html).

```python
def isPrime(x):
    if x<2: return False
    for i in range(2,int(x**0.5)+1):
        if x%i==0:
            return False
    return True

# using a loop
primeList=[]
for i in range(2,31):
  if isPrime(i): primeList.append(i)
print(primeList)

# using a list comprehension
print([i for i in range(30) if isPrime(i)])

# using the filter function
print(list(filter(isPrime,range(0,30))))
```

## Basic Types in Python

- Integer: int
- Floating point: float

There are also more complex types, like lists, tuples, sets and dictionaries, that we will cover later. Python also allows for arbitrary user-defined types, and many modules, like pandas and NumPy, define their own types.

## The Boolean Type

Variables of a boolean type can take on two values, True or False. They can be set as the result of a comparison and can be used as the condition in "if" or "while" statements. Here is an example that uses a temperature.

They can be combined using the "and," "or," and "not" logical operators.

```python
temp=31
freezing=(temp<32)
print(freezing) # True
boiling=(temp>212)
print(boiling) # False
print(type(boiling))
print(boiling or freezing) # True
print(boiling and freezing) # False
print(not(boiling)) # True
```

## Python: Strongly and Dynamically Typed

Strong typing in this context means that there is no automatic conversion of objects to the necessary type. In the following example, Python could have chosen to convert the integer 4 to a string and concatenate it to "hello" to get the string "4hello", but that is not the policy, and therefore it generates a run-time error which is trapped by the "try ... except" syntax. (The "+" operator in the context of strings signifies concatenation.)

```python
try:
  print(4+"hello")
except:
  print("Error generated; can't add different types") 
```

However, it has no problem combining like objects, selecting an appropriate operation based on the types of objects involved:

```python
print(4+6) # addition
print("hello"+" world") # string concatenation
print([1,2,3]+["hello", "world"]) # list concatenation
```

## Dynamic Typing

Python allows you to easily change the type of a variable without any type declarations. It deduces the type based on the context.

In the following example, the variable x is set first to an integer, and then to a string. This works because Python variables are pointers to objects in memory.

```python
# easily change the type of a variable
# x is initially set to a string
x="hello"
print(x)
print(type(x))
# x is now set to an integer
x=5
print(x)
print(type(x))
```

## The "Input" Function and the "If" Statement

Python collects input with the "input" function, and makes decisions with the "if" statement which has optional "elif" and "else" components.

The following collects a temperature from the user and classifies it.

```python
temp=float(input("Enter a temperature in Fahrenheit: "))
if temp<32:
  print("Below the freezing point of water")
elif temp>212:
  print("Above the boiling point of water")
else:
  print("Between boiling and freezing")
```

## Loops: "For" Loops

Python has "for" loops when a fixed number of repetitions are involved.

```python
# add the numbers from 1 to 10 using a for loop
total=0
for i in range(1,11): # the range function here runs from 1 to 10 (11 is omitted)
  total+=i
print(total)
```

You can also loop through a list or a tuple:

```python
for item in ("apple","pear","peach"):
    print(item)
```

## Loops: "While" Loops

"While" loops are condition-controlled. Here is a while loop that adds up the numbers from 1 to 10.

```python
# do the same using a while loop
i=1
total=0
while i<=10:
  total+=i
  i+=1
print(total)
```

## Strings

Strings are sequences of characters. There are various methods available on strings, for instance upper(), lower(), title() and capitalize(). (Methods are functions that are specific to a particular class of objects.)

- upper: converts the string to all upper case
- lower: converts the string to all lower case
- title: capitalizes each word in the string
- capitalize: capitalizes the first latter in the string

For example:

```python
phrase='the rain in Spain is mainly on the plane'
print("The original string is:", phrase)
print("In upper case:", phrase.upper())
print("In lower case:", phrase.lower())
print("In title case:", phrase.title())
print("Capitalized:",phrase.capitalize())
```
