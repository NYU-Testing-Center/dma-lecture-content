## Proprietary Systems for Statistics

SAS, SPSS, and Stata are "old-school" proprietary statistics packages but are still widely-used.

SAS is an enterprise-grade analytics platform widely used in industry and government for large-scale data management, statistical analysis, and reporting. It is command-oriented with its own programming language.

SPSS is a menu-driven statistical package favored in the social sciences for its ease of use in standard statistical tests and survey analysis.

Stata is a programmable statistical environment popular in economics and epidemiology for rigorous data manipulation, reproducible workflows, and advanced econometric analysis. Like SAS, it is command-oriented with its own programming.

All three have developed interfaces that allow their systems to be controlled from Python, but are still proprietary at the core.

## Systems for Numerical Computing, Including Vector Operations

MATLAB is a commercial, proprietary numerical computing environment with an integrated IDE, extensive optimized libraries, and strong vendor support, widely used in engineering, signal processing, and control systems.

Octave is an open-source, MATLAB-compatible language focused on numerical computation, offering similar syntax and core functionality but with fewer toolboxes and a less polished ecosystem.

NumPy is a  open-source Python library for numerical arrays and linear algebra that emphasizes performance and composability, relying on the broader Python scientific stack (e.g., SciPy, pandas, matplotlib) rather than a standalone environment.

Julia is a general-purpose language for numerical computing, emphasizes array programming and linear algebra, interoperates well with other languages, and is designed for high performance without dropping into C/C++. Unlike NumPy: Julia is a full language and environment, not a library layered on top of another language,

## Open Source Options for Data Management and Analysis

Python, R, and Julia are open source. All three rely on optimized native code to carry out frequently-done operations.

Python is a flexible, end-to-end platform that has data science as one of its many applications, whereas R is a specialized tool that focuses on data analysis and statistics

Python and R are both used by data scientists. Python is a general-purpose programming language designed broadly for software development, while R is a domain-specific language designed primarily for statistics and data analysis.

Python’s data science power comes from libraries (NumPy, pandas, SciPy, scikit-learn, PyTorch), whereas R’s strength lies in its vast, statistics-oriented package ecosystem (CRAN, Bioconductor)

By academic discipline:

Statistics, biostatistics, epidemiology, and parts of sociology still lean heavily toward R because of its unmatched depth in classical and modern statistical methods and domain-specific packages.

Computer science, engineering, physics, astronomy, and computational biology favor Python (often alongside C/C++ or Julia).

Interdisciplinary data science and social science increasingly default to Python, especially where ML, NLP, or large datasets are involved.

in this course, we focus on Python.

## "Dashboarding" Options

There are various proprietary systems, like Tableau and Microsoft Power BI, that allow you to build interactive virtual "dashboards" (like the informational dashboard in a car or the control panel in an airplane) for visualizing data. These platforms can be used by non-programmers.

Tableau, the market leader, is a visual analytics and business-intelligence platform that enables users to explore, analyze, and present data through interactive dashboards and charts using a drag-and-drop interface, emphasizing ease of use, fast exploratory analysis, and integration with a wide range of data sources.

There are also open source systems like [Dash](https://plotly.com/examples/), Posit (Shiny), Streamlit and Bokeh that can also be used for developing interactive dashboards.

Dash is an open-source Python framework for building interactive, web-based analytical applications and dashboards directly from Python code, combining Plotly visualizations with reactive UI components without requiring traditional front-end web development.

There are NYU SPS [classes](https://www.sps.nyu.edu/professional-pathways/courses/DATA1-CE9000-visual-analytics-with-tableau.html) on data analytics that focus mainly on Tableau.

## Why Python?

Python has fairly quickly become the largest and most popular data science ecosystem.

R is also quite popular, especially with statisticians. Some packages are available only in R, and some only in Python. But in this course we will be focusing on Python. Many tasks can be done in either language

Typically, people use Python when they need a general-purpose language that spans data analysis, machine learning, engineering, and production systems.

And they use R when their task is primarily statistics, such as experimental design, hypothesis testing, causal inference, or advanced regression modeling, or they are using specialized statistical methods (biostatistics, epidemiology, psychometrics, survey analysis) that are best supported in R.

Python data science applications typically run quickly because often the most compute-intensive aspects of modules have been implemented in C.

## Python vs. R

R is great at working with multidimensional array data. It is also great at holding and manipulating rectangular statistical tables ("dataframes") in memory.

However, Python has been extended to support multidimensional arrays through the NumPy package and dataframes through the pandas package. It now handles these as well as R, using these add-ons.

Arrays in R and in NumPy were influenced by the matrix-centric thinking, vectorized operations, and interactive numerical workflows found in languages developed earlier, like APL, S, and MATLAB.

## Conda and Pip

Conda is the package manager associated with Anaconda. To install Python packages in Anaconda, use the conda command. For instance, the following command installs *scikit-learn*, the most popular machine learning library for Python.

```text
conda install scikit-learn
```

Pip is the main other package manager for Python. It is used in a similar manner. However, the developers of anaconda prefer that you try to use conda first and only use pip if a package isn't available with conda.

You can also use pip to install packages in a Google Colab notebook (into the virtual machine associated with that notebook.) However, this won't persist, so you will usually need to reinstall every time you use the notebook.

## Environments

Conda allows for the creation of environments that contain python and only specific packages. You can also select a particular version of Python or a specific version of a Python package to be run in a particular environment. You probably won't have much need to do this this semester; you can just use the default environment.

If you did want to create an environment called "myenv" the following command would do it.

```text
conda create --name myenv
```

Subsequent commands would install packages into that environment and then activate the environment so you can use it.

More detailed instructions on how to create environments can be found [here](https://docs.conda.io/projects/conda/en/latest/user-guide/tasks/manage-environments.html).

Environments are often used to avoid "dependency hell" where two packages require different versions of a library. If you install the current version or an earlier version of a particular package globally on your system, you can sometimes break the functionality of a package that you are already using. By putting a new package into its own environment, you prevent that from happening.

Python has several other tools for managing dependencies and creating environments, including [virtualenv](https://virtualenv.pypa.io/en/latest/), [uv](https://docs.astral.sh/uv/) and [poetry](https://python-poetry.org/).

Many people use containerization systems, most popularly Docker, to create isolated environments. These are not specific to Python but can be used with it.
