## Indexes in SQLite

This notebook contains a simple demonstration of the performance boost that can be associated with creating an index in SQLite. An index on one or more columns in a database can speed up searching on those columns. Of course, an index is not free; it takes up more space. So we have an example of the familiar space/time tradeoff in computer science.

Another cost of an index is that it takes a little longer to insert items in the affected columns because the index must be updated as well as the columns themselves.

Now we create a query to find all of the numbers between 100 and 110 inclusive. This requires searching  the entire table. We keep track of how long it takes.

Before running the search, we can ask SQLite how it plans to do it using "EXPLAIN QUERY PLAN".

```python
import sqlite3
import pandas as pd
import os
import random
try:
  os.remove("index_test.db") # remove the database if it is already there
  print("removed earlier copy of index_test.db")
except:
  print("not already there")
# create a connection to a database
conn = sqlite3.connect("index_test.db") # creates the database if it is not already there.
def execSQL(conn,query):
  conn.execute(query) # execute an SQL query
  conn.commit() # "commit" that query in order to make its action permanent
def allrowsSelect(conn,query):
  cursor = conn.execute(query)
  for row in cursor:
    print(row)
# put the SQL to create the tables in a string
execSQL(conn,"drop table if exists test")
createCMD='''
CREATE TABLE if not exists test (
  i integer primary key,
  v integer
)
'''
# execute it
execSQL(conn,createCMD)


query='INSERT INTO test (v) VALUES '
n=100000
for i in range(n):
  query+="("+str(random.randint(1,100000))+")"
  if i<n-1: query+=", "
#print("Our query starts with '",query[:100],"' ...",sep='')
execSQL(conn,query)
read=pd.read_sql_query('explain query plan select v from test where v>=100 and v<=110',conn)
print(read)
```
