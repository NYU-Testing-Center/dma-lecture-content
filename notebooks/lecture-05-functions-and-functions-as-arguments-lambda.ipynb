## Return

The return statement does two things

- stops function execution as soon it is called (like break from a loop)
- gives back value
  - If there's no value returned, a function gives back None

```python
def f(a,b):
    result=a*b
print(f(2,3)) # prints None
def f2(a,b):
    return a*b
print(f2(2,3)) # prints 6
```

## "Variadic" Functions (Variable Number of Arguments)

Use \* before an argument name to collect all positional arguments into single parameter:

- parameter name can be whatever you want, but args is a commonly used name
- resulting value of that parameter is a tuple containing all arguments
- if there are additional positional, required arguments, then args must go last
- inspired by other, older Unix tools like the C shell which have a similar mechanism

## An Example of an Arbitrary Number of Arguments

```python
def sum_all(*args):
    print(args)
    print(type(args)) # parameter's type is a tuple; holds all arguments
    total = 0
    for arg in args:
        total += arg
    return total

result = sum_all(1, 2, 3, 4, 5, 6, 10)
print(result)
result = sum_all() # empty tuple if nothing passed in
print(result)
```

## Required Positional Arguments, Then Arbitrary Arguments

```python
def sum_all(a, b, *args):
    print(args)
    print(type(args))
    total = a + b
    for arg in args:
        total += arg
    return total

print(sum_all(1, 2, 3, 4, 5))
print(sum_all(2,3))

# since the first two arguments are required, 
# omitting them will generate an error
try:
    sum_all()
except TypeError as e:
    print('error:', e)
```

## Unpacking an Iterable into Arguments

In the context of a function call, \* breaks up an iterable object, like a list or tuple, into separate arguments.

```python
fruits=['apple','banana','orange']
print(fruits)
print(*fruits)
print('apple','banana','orange')

def sum_all(a, b, *args):
    print(args)
    print(type(args))
    total = a + b
    for arg in args:
        total += arg
    return total

nines=[9,9,9]
print(sum_all(1,2,*nines))
print(sum_all(1,2,9,9,9))

# without unpacking, generates an error
try:
    sum_all(1,2,nines)
except TypeError as e:
    print('error:', e)
```

## Default Values/Keyword Arguments

Keyword arguments can be given in any order and have default values.

```python
# n is the default value, but can be set to something else
def shout(n=3):
    return 'hello' + n * '!'
print(shout())
print(shout(n=30))
```

```python
def greet(fn="John",mn="Quincy",ln="Adams"):
    print("Hello", fn)
    print("Your middle name is", mn)
    print("Your full name is",fn,mn,ln)
    return
greet()
greet(mn="Earl",ln="Jones",fn="James") # can use other order
# if names aren't specified uses order
greet("William","Jefferson","Clinton")
```

## An Arbitrary Number of Keyword Arguments

are collected into a dictionary.

```python
def g(**kwargs):
    print(kwargs)
g(a=1, b=2, c=3)
```

```python
def greet2(**kwargs):
    print("Hello", kwargs["fn"])
    print("Your middle name is", kwargs["mn"])
    print("Your full name is",kwargs["fn"],kwargs["mn"],kwargs["ln"])
    return
greet2(fn="James",mn="Earl",ln="Jones")
greet2(mn="Earl",fn="James",ln="Jones",misc="Foo")

try:
  greet2(mn="Earl")
except:
  print("missing some arguments")
```

## Filter: Apply a Boolean Filter to a Sequence

The filter function takes two arguments:

1. A function (which may be a lambda) which returns a Boolean value (True/False)
2. A sequence

It returns a sequence that contains just the elements of the original sequence for which the function returns true.

Here are some examples:

```python
# start with the numbers between 1 and 6
l=list(range(1,7))
print(l)

# find the odd numbers
odd=filter(lambda n: (n%2)==1, l) # returns filter object
print(list(odd))

# find the even numbers
even=filter(lambda n: (n%2)==0, l) # returns filter object
print(list(even))

#two ways to use a named function
def isOdd(n): return (n%2)==1
isOdd2=lambda n: (n%2)==1

# either one gives the same result
print(list(filter(isOdd,l)))
print(list(filter(isOdd2,l)))
```

## Reduce: Repeatedly Apply a Function to a Sequence

The reduce function takes two arguments:

1. A function of two arguments
2. A sequence

It returns the result of repeatedly applying the function to the sequence. Note it is not part of standard Python and needs to be imported from the functools module.

Here's our version of the sum function:

```python
from functools import reduce
oursum= lambda l: reduce(lambda x,y: x+y,l) # repeatedly adds up elements in l

print(oursum([1,2,3,4]))
print(sum([1,2,3,4])) # built in version

# version with variable number of arguments
oursum2= lambda *l: reduce(lambda x,y: x+y,l)
print(oursum2(1,2,3,4))
```
