```python
import sqlite3
import pandas as pd
import os
try:
  os.remove("artists_tracks.db") # remove the database if it is already there
  print("removed earlier copy of artists_tracks.db")
except:
  print("database not already there")
# create a connection to a database
conn = sqlite3.connect("artists_tracks.db") # creates the database if it is not already there.
def execSQL(conn,query):
  conn.execute(query) # execute an SQL query
  conn.commit() # "commit" that query in order to make its action permanent
def allrowsSelect(conn,query):
  cursor = conn.execute(query)
  for row in cursor:
   print(row)

# create a connection to a database
conn = sqlite3.connect("artists_tracks.db") # creates the database if it is not already there.
execSQL(conn,"PRAGMA foreign_keys=ON")
# put the SQL to create the tables in a string
createCMD='''
CREATE TABLE IF NOT EXISTS artist (
  artistid    INTEGER PRIMARY KEY,
  artistname  TEXT
)
'''
# execute it
execSQL(conn,createCMD)
```

The code below creates the tracks table. Note that the foreign key statement adds the constraint that the foreign key that points back to the artists table must be present in the artists table.

```python
import sqlite3
import pandas as pd
import os
try:
  os.remove("artists_tracks.db") # remove the database if it is already there
  print("removed earlier copy of artists_tracks.db")
except:
  print("database not already there")
# create a connection to a database
conn = sqlite3.connect("artists_tracks.db") # creates the database if it is not already there.
def execSQL(conn,query):
  conn.execute(query) # execute an SQL query
  conn.commit() # "commit" that query in order to make its action permanent
def allrowsSelect(conn,query):
  cursor = conn.execute(query)
  for row in cursor:
   print(row)

# create a connection to a database
conn = sqlite3.connect("artists_tracks.db") # creates the database if it is not already there.
execSQL(conn,"PRAGMA foreign_keys=ON")
# put the SQL to create the tables in a string
createCMD='''
CREATE TABLE artist(
  artistid    INTEGER PRIMARY KEY,
  artistname  TEXT
)
'''
# execute it
execSQL(conn,createCMD)
query='''
INSERT INTO artist (artistid, artistname)
	VALUES
	  (1, 'Dean Martin'),
	  (2, 'Frank Sinatra')
'''
execSQL(conn,query)
createCMD='''
CREATE TABLE track(
  trackid     INTEGER PRIMARY KEY,
  trackname   TEXT,
  trackartist INTEGER,
  FOREIGN KEY(trackartist) REFERENCES artist(artistid)
)
'''
# execute it
execSQL(conn,createCMD)
```
