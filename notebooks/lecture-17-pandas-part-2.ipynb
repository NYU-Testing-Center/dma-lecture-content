Using the corporate data, here we see how we can use the loc method and a list of columns to just select certain rows and columns from a DataFrame.

```python
import pandas as pd
d = [["$229.2", 2017, 123000, "$1100", "Cupertino, US"],
     ["$211.9", 2017, 320671, "$284", "Suwon, South Korea"],
     ["$177.8", 2017, 566000, "$985",  "Seattle, US"],
     ["$154.7", 2017, 1300000, "$66", "New Taipei City, Taiwan"],
     ["$110.8", 2017, 80110, "$834", "Mountain View, US"]]

comps = ["apple", "samsung", "amazon", "foxconn", "alphabet"]
cols = ["revenue", "fy", "employees", "mcap", "location"]

c = pd.DataFrame(d, index=comps, columns=cols)
# here's how we look at just some of the data
print(c.loc[['alphabet','amazon'],['location','revenue']])
```

```python
import pandas as pd
d = [["$229.2", 2017, 123000, "$1100", "Cupertino, US"],
     ["$211.9", 2017, 320671, "$284", "Suwon, South Korea"],
     ["$177.8", 2017, 566000, "$985",  "Seattle, US"],
     ["$154.7", 2017, 1300000, "$66", "New Taipei City, Taiwan"],
     ["$110.8", 2017, 80110, "$834", "Mountain View, US"]]

comps = ["apple", "samsung", "amazon", "foxconn", "alphabet"]
cols = ["revenue", "fy", "employees", "mcap", "location"]

c = pd.DataFrame(d, index=comps, columns=cols)

del c['mcap']
print(c)
```

And let's remove the fy column using the drop method.

```python
import pandas as pd
d = [["$229.2", 2017, 123000, "$1100", "Cupertino, US"],
     ["$211.9", 2017, 320671, "$284", "Suwon, South Korea"],
     ["$177.8", 2017, 566000, "$985",  "Seattle, US"],
     ["$154.7", 2017, 1300000, "$66", "New Taipei City, Taiwan"],
     ["$110.8", 2017, 80110, "$834", "Mountain View, US"]]

comps = ["apple", "samsung", "amazon", "foxconn", "alphabet"]
cols = ["revenue", "fy", "employees", "mcap", "location"]

c = pd.DataFrame(d, index=comps, columns=cols)
del c['mcap']
# axis 1 is columns
# inplace is destructive (by default it makes a copy since the default value of inplace is False)
c.drop('fy',axis=1,inplace=True)
print(c)
```

What if we want to just see the employees column? Note that it is a Pandas Series. We can also easily convert it to a list or a dictionary if we like.

```python
import pandas as pd
d = [["$229.2", 2017, 123000, "$1100", "Cupertino, US"],
     ["$211.9", 2017, 320671, "$284", "Suwon, South Korea"],
     ["$177.8", 2017, 566000, "$985",  "Seattle, US"],
     ["$154.7", 2017, 1300000, "$66", "New Taipei City, Taiwan"],
     ["$110.8", 2017, 80110, "$834", "Mountain View, US"]]

comps = ["apple", "samsung", "amazon", "foxconn", "alphabet"]
cols = ["revenue", "fy", "employees", "mcap", "location"]

c = pd.DataFrame(d, index=comps, columns=cols)

print(type(c['employees']))
number_of_employees=list(c['employees'])
print(c['employees'])
print(number_of_employees)
number_of_employees_d=dict(c['employees'])
print(number_of_employees_d)
```

```python
#@title Solution
import pandas as pd
d = [["$229.2", 2017, 123000, "$1100", "Cupertino, US"],
     ["$211.9", 2017, 320671, "$284", "Suwon, South Korea"],
     ["$177.8", 2017, 566000, "$985",  "Seattle, US"],
     ["$154.7", 2017, 1300000, "$66", "New Taipei City, Taiwan"],
     ["$110.8", 2017, 80110, "$834", "Mountain View, US"]]

comps = ["apple", "samsung", "amazon", "foxconn", "alphabet"]
cols = ["revenue", "fy", "employees", "mcap", "location"]

c = pd.DataFrame(d, index=comps, columns=cols)

c['state'] = pd.Series({'apple': 'CA', 'amazon': 'WA', 'alphabet': 'CA'})

geosplit=c['location'].str.split(',')
c['country']=geosplit.str[-1]
c['city']=geosplit.str[0]
print(c)
# note the use of * here to unpack the elements of a list
# they can be passed to the reindex method.
c=c.reindex(index=[*(list(c.index)[1:]), 'apple', 'microsoft'],
            columns=['city','state','country','revenue', 'employees'])
print(c)
```

Here we use iloc in combination with a list to fill in the remaining data for Microsoft.

The line of code `c.iloc[5,2:]=["US",161,182268]`, it updates the row at index position 5 (the row for Microsoft), starting from the third column (since Python is zero-indexed, column indexes start at 0, so index 2 is the third column) and sets the 'country' column to `"US"`, the 'revenue' column to `161`, and the 'employees' column to `182268`.

```python
#@title Solution
import pandas as pd
d = [["$229.2", 2017, 123000, "$1100", "Cupertino, US"],
     ["$211.9", 2017, 320671, "$284", "Suwon, South Korea"],
     ["$177.8", 2017, 566000, "$985",  "Seattle, US"],
     ["$154.7", 2017, 1300000, "$66", "New Taipei City, Taiwan"],
     ["$110.8", 2017, 80110, "$834", "Mountain View, US"]]

comps = ["apple", "samsung", "amazon", "foxconn", "alphabet"]
cols = ["revenue", "fy", "employees", "mcap", "location"]

c = pd.DataFrame(d, index=comps, columns=cols)

c.drop('fy',axis=1,inplace=True)
number_of_employees=list(c['employees'])
c['state'] = pd.Series({'apple': 'CA', 'amazon': 'WA', 'alphabet': 'CA'})
geosplit=c['location'].str.split(',')
c['country']=geosplit.str[-1]
c['city']=geosplit.str[0]

c=c.reindex(index=[*(list(c.index)[1:]), 'apple', 'microsoft'],
        columns=['city','state','country','revenue', 'employees'])

c.loc['microsoft','city']='Seattle'

c.iloc[5,1]="WA"
c.iloc[5,2:]=["US",161,182268]
print(c)
```
