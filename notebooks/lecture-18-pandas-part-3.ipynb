Apply allows a function to be called on a ow or column.

d.apply(fn,axis=a)

- fn is the function name
- axis may be 0 for rows (default) or 1 for columns

In the code below, we use apply to find the total rainfall per month across all four cities. The result, since it is one dimensional, is a series.

Since axis is 0, it sums across the rows for each column.

```python
# rain data
import pandas as pd
rain = pd.DataFrame([[3.50, 4.53, 4.13, 3.98],
                     [7.91, 5.98, 6.10, 5.12],
                     [3.94, 5.28, 3.90, 4.49],
                     [1.42, 0.63, 0.75, 1.65]],
    index=['New York', 'New Orleans', 'Atlanta', 'Seattle'],
    columns=['Jun', 'Jul', 'Aug', 'Sept'])
print(rain)
print()
total_rain_by_month=rain.apply(lambda x: sum(x)) # if axis is omitted, it is set to zero (rows)
print(total_rain_by_month)
print(type(total_rain_by_month))
```

The code below defines  a function to compute the factorial of a number and applies this function to a 3x3 DataFrame generated from a sequence of numbers.

```python
import numpy as np
import pandas as pd

# apply factorial to the numbers in a 3x3 DataFrame containing the numbers from 0 to 8
def factorial(n): return 1 if n==0 else n*factorial(n-1)
nums=pd.DataFrame(np.arange(9).reshape(3,3))
print(nums)

nums=nums.map(factorial)
print(nums)
```

Here we load the company data again, and then reformat the revenue column by removing the dollar sign and converting the string to a float.

Note we couldn't run describe on that column without turning it into a number.

```python
import pandas as pd
d = [["$229.2", 2017, 123000, "$1100", "Cupertino, US"],
     ["$211.9", 2017, 320671, "$284", "Suwon, South Korea"],
     ["$177.8", 2017, 566000, "$985",  "Seattle, US"],
     ["$154.7", 2017, 1300000, "$66", "New Taipei City, Taiwan"],
     ["$110.8", 2017, 80110, "$834", "Mountain View, US"]]

comps = ["apple", "samsung", "amazon", "foxconn", "alphabet"]
cols = ["revenue", "fy", "employees", "mcap", "location"]

c = pd.DataFrame(d, index=comps, columns=cols)
# that didn't reset it: the following will
c['revenue']=c['revenue'].map(lambda x: float(x[1:]))
print(c)
print(c['revenue'].describe())
```
