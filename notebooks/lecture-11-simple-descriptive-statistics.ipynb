## Descriptive Statistics

Descriptive statistics about some data include:

1. The mean (simple average)
2. The median (middle value when sorted, or the average of the two middle values)
3. The maximum
4. The minimum

The code below uses a generator to read the calories column from the Starbucks data file. The built-in max and min functions can actually take a generator as an argument, so the code prints the maximum and minimum calories.

A generator is not a collection of elements, so you have to coerce it to a list to pass it to len in order to determine how many calorie values are found in the file and the mean (arithmetic average) of them, which is just their sum divided by the number of values.

We also import the Counter function to create a table of calorie counts.

```python
def get_calories():
    with open('starbucks_drinkMenu_expanded.csv', 'r') as f:
        next(f)
        for line in f:
            line_parts = line.split(',')
            yield int(line_parts[3])

def mean(l):
    return sum(l)/len(l)
print("Maximum:",max(get_calories()))
print("Minimum:",min(get_calories()))
calories=list(get_calories())
print("Number of values:",len(calories))
print(calories[:10]) # look at the first ten values.
print(calories[-10:]) # and the last ten values
print("Mean:",mean(calories)) # print the mean

from collections import Counter
print(Counter(calories))
```

The code below reads the file in with pandas and uses it to get descriptive statistics on the calories column.

The code "std" means standard deviation. There are also the 25th, 50th, and 75th percentiles output, as well as the mean, count, minimum, and maximum. The 50th percentile is the same as the median.

We can also use pandas to get the counts of each value using the value\_counts() method.

```python
import pandas as pd
starbucks=pd.read_csv('starbucks_drinkMenu_expanded.csv')
print(starbucks.describe()[["Calories"]]) # get descriptive statistics for the Calories column
print(starbucks['Calories'].value_counts()) # get the counts for each value of calories.
```

## Frequencies

The frequencies associated with a variable are the number of times each value appears in the dataset.

We can use a dictionary or a Counter object to find this.

Let's set up 1000 rolls of a pair of six-sided dice using NumPy and then count them using a dictionary.

```python
import numpy as np

# roll two six-sided dice
rolls = np.random.randint(1, 7, 1000) + np.random.randint(1, 7, 1000)
# first 20
print(rolls[:20])
counts = {}
for n in rolls:
   counts[n] = counts.get(n, 0)  + 1 # this doesn't fail if the dictionary key is not there

print(counts)
```

## Using Counter

Here is the version using a Counter. We can use it to find the mode, which is the most common value.

```python
import numpy as np

# roll two six-sided dice
rolls = np.random.randint(1, 7, 1000) + np.random.randint(1, 7, 1000)
# first 20
print(rolls[:20])
from collections import Counter
c=Counter(rolls)
print(c)
print(c.most_common(1)) # find the most common
print(c.most_common(1)[0][0]) # just the value
print(c.most_common(2)) # find the two most common
print(c.most_common(4)) # find the four most common
```

## Quantiles

A quantile is a value  under which some given percent of ordered data falls. (Common quantiles in use include percentiles, tertiles, quartiles, etc. ...)

```python
import statistics 
quantile = lambda p, data: sorted(data)[int(p * len(data))-1]
nums = [22, 12, 1, 5, 6, 2, 3, 20, 12, 8]
nums.sort()
print(len(nums))
print(nums)
print(statistics.median(nums)) # the median is the 50th percentile
print(quantile(0.50,nums)) # we truncated, but close
```
