## MongoDB

MongoDB is an example of a document database. Rather than storing data in a collection of tables as is done in a relational database, MongoDB focuses on storing and querying documents in a format based on JSON.

## Relational Databases

#### Relational Databases

- relational databases are typically pretty rigid:
  - highly structured
  - you have to define the columns and the types of columns before inserting rows
  - have a lot of features for maintaining data integrity (such user defined data constraints, foreign keys, etc.)
- Many relational databases guarantee that transactions (or changes in the database) are reliable
- Relational databases often offer ACID compliance - Atomicity, Consistency, Isolation, Durability

## NoSQL Databases

Let's contrast relational databases with NoSQL databases (that is, non SQL, non relational, or even Not Only SQL.)

So, some common ideas (not necessarily correct) about NoSQL Databases, such as key-value and document stores, are that they are:

- sometimes more simple in design and architecture
- are less rigid / more flexible than relational databases
- may be easier to scale "horizontally"
- data is not necessarily normalized
- the underlying structures aren't limited to tables (key-value, document, graph, etc.)

## Compromises Made by NoSQL Databases

NoSQL databases sound good: why didn't we start using them in the first place?

Some compromises that a NoSQL database may have to make:

- not usually ACID compliant (though MongoDB has had some level of ACID compliance since ~2018/ version 4)
- simple data structures and architecture moves constraints and referential integrity to application layer
- lack of standard (or at least mostly standard) language like SQL
- some NoSQL systems even exhibit lost writes or data loss (because not ACID compliant)
- Again, these are very broad generalizations (some NoSQL databases are ACID compliant)

## Document Stores

In document stores, data is organized (as the name implies) as semi-structured documents

- think JSON (other formats too: XML, YAML, etc.)
- or… a richer key-value store (there's meta data within the document… the keys are usually meaningful)
- typically, no schema is required (that is, data types of values are inferred from values)
- typically, semi structured (documents, property names, etc… do not have to be pre-defined with schemas)
- some document stores are known for high availability and scaling through replication / redundancy and sharding (separating large databases into smaller ones)

They are good for applications where flexible data storage or constantly changing data storage is required.

## MongoDB

MongoDB is a Nosql database. Specifically, it's a document store:

- a single record in Mongo is a document
- a document is a bunch of key value pairs…
- documents are similar to JSON / JavaScript objects or Python dictionaries
- it's actually BSON, a binary version of JSON
- for convenience, we can say that MongoDB uses JSON document to store records

## JSON

JSON (JavaScript Object Notation) is a data interchange format / file format that is composed of key value pairs.

A single JSON document consists of:

- curly braces
- keys and values joined by :
- key/value pairs separated by commas
- all keys must be double quoted
- values can be numbers, strings, arrays or other JSON documents/objects
- (you can have documents embedded/nested within other documents)

## BSON

MongoDB uses a binary-encoded format to store JSON documents

this format is called BSON or binary JSON

it's similar to JSON (it has embedded documents, arrays, etc.), but…

- it has additional data types
- more space efficient
- it is faster to read from or write to the disk or memory

## Documents and Collections

Terminology:

- key - a field name - analogous to a column in a relational database
- value
- document - a single object or record in our database
  - consists of key value pairs
  - similar to a single row in a relational database
- collection - a group of documents
  - analogous to a table in a relational databases

## MongoDB

The open source version of MongoDB is called the Community Edition. You can run it in the cloud or on your local machine.

## Some mongosh commands

The following commands can be used to navigate, create and remove databases and collections

- show databases - show available databases (remember, there can be more than one database)
- use db - work with a specific database (if unspecified, the default database connected to is test)
- show collections - once a db is selected, show the collections within the database
- db.dropDatabase() - drop (remove) the database that you're currently in (must use first)
- db.collectionName.drop() - drop (remove) the collection named collectionName
- To get some inline help:
  - help - get help on available commands

## MongoDB Queries

MongoDB uses a [query-by-example](https://en.wikipedia.org/wiki/Query_by_Example) style of querying. Query by Example is a technique that dates back to the 1970s. That means that you give MongoDB a pattern that you want to see in your results, and it returns the results that match your pattern. Pattern matching via [regular expressions](https://www.mongodb.com/docs/manual/reference/operator/query/regex/) is supported.

## Performance and Suitability

MongoDB claims performance advantages. But these are refuted by others. So it is hard to say which is true. I'm skeptical of any performance advantages, especially because all systems are moving targets and some of the studies aren't impartial.

OnGres [found](https://www.enterprisedb.com/news/new-benchmarks-show-postgres-dominating-mongodb-varied-workloads) that PostgreSQL was faster than MongoDB, but MongoDB [disputes this](https://www.mongodb.com/blog/post/benchmarking-do-it-right-or-dont-do-it-at-all), saying that OnGres's methodology was poor. And then OnGres [critiqued](https://www.ongres.com/blog/benchmarking-do-it-with-transparency/) MongoDB's critique.

The debate is not only over performance, but over the suitability for usage as a database at all. This [article](https://www.enterprisedb.com/blog/schema-later-considered-harmful) by Michael Stonebraker illustrates how the lack of a predefined schema can be a disadvantage. And this [article](https://www.enterprisedb.com/blog/compare-mongodb-vs-postgresql-join-command), also by Stonebraker, critiques MongoDB's join facilities.

Google Spanner is an example of a relational system that scales comparably well to MongoDB and similar NoSQL systems.

```python
db.person.delete_many({})
# now both records are gone
for i in db.person.find({}):
  print(i)
```

## Alternatives to persisting pandas data in a database

Python offers its own facilities for object persistence including "shelve" and "pickle" which are more lightweight than persisting data to a MongoDB collection or a table in some SQL database.

```python
import shelve
enrollments_persist=shelve.open("enrollments") # this creates a file called "enrollments.db"
for key in enrollment_dict.keys():
    enrollments_persist[key]=enrollment_dict[key]
enrollments_persist.close()
```

Shelve files are key-value stores like dictionaries. Here we re-open the persistent file and print out one value.

```python
e2=shelve.open("enrollments")
print(e2["CSCI-UA-0201-003 mt6271"])
e2.close()
```

Pickle can serialize (convert to a byte stream and write to disk) any Python object. Unpickling can be insecure, so only unpickle trusted objects. Pickling is built into pandas.

```python
enrollments.to_pickle("enrollments.pickle")
```

```python
e3=pd.read_pickle("enrollments.pickle")
e3.head(2)
```

Another option for persisting data is to just convert it to JSON and write it out to a text file.

```python
f=open("enrollments.json","w")
f.write(enrollments.to_json(orient='index'))
f.close()
```

```python
f=open("enrollments2.json","w")
f.write(enrollments.to_json(orient='records',lines=True))
f.close()
```
