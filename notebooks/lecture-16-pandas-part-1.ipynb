## Pandas

Pandas ("Python data analysis") is a Python module that contains data structures and tools for working with data.

It has a lot of numpy-like functionality in it, like vector-based operations

However, pandas specializes in dealing with tabular data composed of mixed data types, while an numpy array can usually hold a single data type (although there is a newer type of numpy array, a structured array, that can hold more than one.)

Also, pandas has an explicit idea of named columns and an index column that indexes the rows. Numpy just has indeces and slices (which pandas has as well.)

## Types in Pandas

- Series: a one-dimensional, labeled array
- DataFrame: two dimensional data structure (with named columns, like a spreadsheet with a header row)
  - each column of a DataFrame is itself a Series
- Index

## Pandas Series

- a one-dimensional labeled array
- the associated labels are collectively referred to as the index
- Series properties:
  - values: the actual data in the Series
  - index: the labels for the data in the Series

There are several ways to create a series

- use a single positional argument, which may be a ndarray or a Python sequence like a list, to specify the series's values.
- two positional arguments, with the second containing the index labels
- passing in keyword arguments for data and index
- passing in a dictionary with the keys as labels and the values as the series' values

In the example below, we initialize a pandas series with a list.

```python
import pandas as pd
import numpy as np
# without the index specified, you get numerical index values starting with zero
# the index is displayed in the first column and the values in the second
series =pd.Series([7,8,9])

print(series)
```

We can set a custom index, rather than the default numeric one. In the example below, each state is indexed by its two-letter postal abbreviation (a string).

Note that the index is not unique in this case. It doesn't have to be. In NumPy, indexes must be numeric and unique.

```python
import pandas as pd
import numpy as np
# and unlike numpy, labels can be non-numeric and even non-unique
cities=pd.Series(["Stamford","Albany","Buffalo","Hartford"],["CT","NY","NY","CT"])
print(cities)
print()
print(cities['NY'])
```

Note that if an index key is specified but the data is not, the data value becomes NaN (which stands for "Not a Number" and is one of [the missing value representations in Pandas](https://pandas.pydata.org/docs/user_guide/missing_data.html).) Missing data representations allow us to specify that we don't know what the value is. Note this can mean two things: that we don't know what the value is, or that there is no value in existence. For instance, there is a King of England, but maybe we don't know their name. But there is no King of the United States ...

This is shown below in that there is no value associated with "C" as the value of the index.

The "isnull" method checks if series values are null; the notnull method checks if they are not.

```python
import pandas as pd
import numpy as np
# if an index label is not in the dictionary, it is included as NaN missing data.
x=pd.Series({'A':'ant','B':'bat'},['A','C'])
print(x)
print()
print(pd.isnull(x))
print()
print(pd.notnull(x))
```

  </div>

The example extracts those rows where the value of the first column (cA) is not 2.

The expression translates to a boolean vector.

As we will see, this is similar to the following SQL:

select \* from d where cA!=2

where d is a table in an SQL-based relational database.

```python
import pandas as pd
d = pd.DataFrame({"cA": {'r1': 1, 'r2': 2, 'r3': 3},
                  "cB": {'r1': 4, 'r2': 5, 'r3': 6},
                  "cC": {'r1': 7, 'r2': 8, 'r3': 9}})
print("Original Dataframe:")
print(d)
# or using an expression
# give me the rows where the value in column cA is not 2
# similar to SQL "select * from d where cA!=2"
print("Conditional Produces Boolean Vector:")
print(d['cA']!=2)
print("Extracted Dataframe")
print(d[d['cA']!=2])
```

In the example below, we use the drop method to remove a column from a dataframe.

The drop method can drop either columns or rows based on the value of the axis parameter, where axis=0 means rows and axis=1 means columns.

Unless you specify inplace=True, the method will create a copy. If you do specify inplace=True, it will destructively modify the existing object.

```python
import pandas as pd
# can add columns
df = pd.DataFrame([[4, 5, 6], [7, 8, 9]],
     columns=['foo', 'bar', 'baz'])
df['qux'] = [20, 30]  # qux is new
print(f"Original Dataframe:\n{df}")


# dropping columns
# need to specify axis
# rows is axis 0
# columns is axis 1
# need to specify inplace if you don't want to just create a copy
# with the column dropped but want to destructively remove the column
#df=df.drop('bar',axis=1)
df.drop('bar',axis=1,inplace=True)
print(f"Modified Dataframe with Column Removed:\n{df}")
```
