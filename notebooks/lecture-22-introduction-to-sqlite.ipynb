## SQLite

[SQLite](https://www.sqlite.org/) is a popular "embedded" database which does not use the client-server model but rather stores its data in an ordinary file on a computer. It is heavily used by many operating systems and applications. It supports much of the SQL standard. It is not an application but a C-language library which other developers can use in their apps.

It has an amusing slogan: "Small. Fast. Reliable. Choose any three."

I'm using it in this course because it is simple but features almost all of the significant features of SQL. SQLite is in heavy use in industry, so it is useful to have a knowledge of it; it not just a toy database.

## SQLite Data Types

SQLite supports five basic data types to store its column values:

- NULL: for representing missing data
- INTEGER: uses a variable amount of bytes depending on the value's magnitude
- REAL: stores a floating point value in eight bytes
- TEXT: a text string
- BLOB: A "blob" of data (binary large object) which can be anything, stored just as it was input. Can be used to store images, for instance.

## Creating Databases

In other relational databases that use the client server model, like PostgreSQL or MySQL, databases are created using the "CREATE DATABASE" command. Each database server can host multiple databases. In SQLite, databases can be created at a command prompt, e.g.

```text
sqlite3 test.db
```

creates a database called "test.db" which is stored in an ordinary system file which can be sent to other machines or deleted. Switching databases is just a matter of opening a different database file.

## Creating Tables in a Database

Every database needs to have one or more tables in it. You can think of each table as similar to a pandas dataframe. It has one or more named columns. The columns can take on any of the types given above.

In SQL, creating tables is done by giving a list of the names of the columns of the tables along with their types (one of the five above) in a CREATE TABLE statement. Some of the information that may be included in this statement is:

- the name of the new table (required)
- the name of each column in the table (required)
- the type of each column (required)
- a default value for a column (optional)
- a PRIMARY KEY for the table, which can be in one or more columns (optional, but a good idea)
- constraints on values (e.g. UNIQUE or NOT NULL) (optional, but often also a good idea)

## The SELECT Statement: Retrieving Data in SQL

The command to query data in SQL is SELECT. The basic structure of a SELECT command is as follows:

```text
SELECT [DISTINCT] columns
  FROM table
  WHERE condition
  GROUP BY columns
  HAVING condtion
  ORDER BY column [ASC | DESC]
  LIMIT offset, count
```

SELECT and FROM are required; all of the rest are optional.

The SELECT command used in the previous command is the simplest one, SELECT \* FROM MOVIE; it selects all of the columns and rows from the MOVIE table, the entire table.

The query below selects just the movies with runtime greater than 125 and then averages up their budget by director, and then restricts using HAVING to those rows with an average budget of over $20 million, ordering the output by the director's name in ascending order.

````python
import sqlite3
import pandas as pd
conn=sqlite3.connect("movies2.db")
query="""
select director, avg(budget) as avg_budget
from movie
where runtime>125
group by director
having avg_budget>20000000
order by director asc
"""
print(pd.read_sql_query(query, conn))
````

## SELECT: Order of Operations

The SELECT command executes in the following order:

1. From: selects the table(s)
2. Where (if any) : applies conditions
3. Group by (if any): creates groups
4. Aggregations (if any): aggregates by groups, typically using a function like avg or max
5. Having (if any): conditions on the aggregated results
6. Select: select columns

## Introducing Joins in SQLite

A join merges data from two or more tables. This can be done in several ways:

- Cross Join: Merges every row in the first table with every row in the second. Rarely Used. Doesn't match values between tables.

The following types of joins match on some common data and merge based on that.

- Inner Join: Includes only rows found in both tables from both tables (intersection)
- Left Outer Join: Includes all of the rows in the left table and matching rows in the right table.
- Right Outer Join: Includes the matching rows in the left table and all rows in the right table. Not supported by SQLite.
- Full Outer Join: Includes all the rows in both the left and right tables whether they match or not. Not supported by SQLite.

A left outer join (also called just a left join) includes all the rows in the left table along with the matching rows in the right table. If no row matches on the right, null values are inserted.

There are also right outer joins (which include everything in the right table) and full outer joins (which include everything in both tables) but these are not supported by SQLite. There are workarounds to this, though.

For instance, you can make a right outer join into a left one by just swapping which tables are left and right.

The following is a left outer join that includes all the rows in the left table (movie), even those without a matching genre id. The right table is genre.

````python
import pandas as pd
import sqlite3
def allrowsSelect(query, conn):
  cursor = conn.execute(query)
  for row in cursor:
    print(row)
conn=sqlite3.connect("movies2_with_genre.db")
query="""
SELECT title, name as genre_name
  FROM movie
  LEFT JOIN genre on movie.genre_id = genre.genre_id
"""
print(pd.read_sql_query(query,conn))
conn.close()
````

    Full Outer Joins
  </div>
</div>

Full outer joins include all the rows in both tables, with those rows with matching keys joined. Where there is no match, NULLs are inserted.

If we have tables A and B, then the full outer join is the union of:

- a left outer join of A (left) with B (right), which includes all the records of A including those matching on the key (and those that don't)
- a partial left outer join of B (left) with A (right) which only includes those records in B which don't match a record in A. (This is the same as a partial right outer join of A and B.)

if A is the Khans table and B is the Khagans table, we can do that with the query below. Note in this case the second left outer join is in fact empty, since as we have seen there are no Khagans that don't match a record in the Khan table, but that doesn't have to be the case with data more generally.

````python
import pandas as pd
import sqlite3
conn=sqlite3.connect("khans.db")
query='''
SELECT * FROM
khans LEFT JOIN khagans
ON khans.khan_id=khagans.khan_id
UNION ALL
SELECT * FROM
khagans LEFT JOIN khans
ON khans.khan_id=khagans.khan_id
WHERE khans.khan_id IS NULL
'''
print(pd.read_sql_query(query,conn))
conn.close()
````

Self or reflexive joins are joins of a table on itself. Some cases where you might want to do this:

- a table of employees that shows employee manager relationships, where the managers themselves are also in the employee table (and have their own manager, except for the CEO.)
- a table of people who might have a spouse in the same table

In our table of Khans, each Khan's parents and children are often found in the same table.

The query below produces a list of all the fathers and their children. Each father has multiple rows if he has more than one child.

Note that we create aliases for the columns and the tables to distinguish between the two copies of the same table.

````python
import pandas as pd
import sqlite3
conn=sqlite3.connect("khans.db")
query='''
SELECT parents.name AS father, children.name as child
FROM khans AS parents INNER JOIN khans AS children
ON parents.khan_id = children.father_id
order by father
'''
print(pd.read_sql_query(query,conn))
conn.close()
````

What if we want to find the list of children of Khans that were also Khagans? This requires a self join on the Khans table combined with an inner join on the Khagans table.

````python
import pandas as pd
import sqlite3
conn=sqlite3.connect("khans.db")
query='''
SELECT parents.name AS father,
GROUP_CONCAT(children.name) AS children,
COUNT(children.name) AS num_kids
FROM khagans, khans as parents, khans as children
WHERE parents.khan_id = children.father_id
AND parents.khan_id=khagans.khan_id
GROUP BY father
ORDER BY num_kids DESC
'''
print(pd.read_sql_query(query,conn))
conn.close()
````

Having found the total urban population of each country, we can find the share that lives in one of the large cities in the list.

This involves a subquery. We will be seeming more examples of subqueries soon. Note the subquery is given a temporary.

````python
import pandas as pd
import sqlite3
conn=sqlite3.connect("cc.db")
query="""
select countries.country,population, cityList, urbanpop,
urbanpop/population as largeCityShare
from (select country, group_concat(City) as cityList,
total(MetroPopulation2018) as urbanpop
from cityPopulations
group by country) as totCityPop, countries
where totCityPop.Country=countries.country
order by largeCityShare desc
"""
print(pd.read_sql_query(query,conn))
conn.close()
````
