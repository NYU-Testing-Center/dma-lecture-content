## Compound Types

In Python, there are many types that contain of a collection of values. Some examples are:

- sequences
  - lists (ordered collection of mutable (changeable) values)
  - tuples (ordered collection of immutable (non-changeable) values)
  - range objects (which represent an arithmetic sequence with a start, stop, and step)
    - E.g. range(3,9,2) generates the numbers 3, 5, and 7
  - strings (ordered collection of characters)
- dictionaries: based on hashed key/value pairs
- sets

```python
print(list(range(3,9,2)))
```

## Common Sequence Operations

- map a function across sequence elements
- iterate (for loop)
- index ([])
- built-in functions: len
- concatenation and repetition: + and \*
- membership: in and not in
- slicing [start:stop:step]

## Indexing

- Each element in a sequence has a numerical position
- The position is called an index
- Indexes start at 0
- Trying to index into a sequence at a position that doesn't exist causes a runtime error

```python
line_one="Let me not to the marriage of true minds admit impediments"
print(len(line_one)) # length is 58
print(line_one[0]) # first character
print(line_one[4]) # letter "m" in "me" 
print(line_one[57]) # last character final index is 57
# the following is out of range
try:
    print(line_one[58])
except IndexError:
    print("Out of range.")
```

## Slicing

With slicing, you specify the starting index (inclusive) and the ending index (exclusive) and an optional skipping value (step), separated by colons, and you get a subsequence. It can be used with any sequence type (list, tuple, or string.)

You can omit one or more of start, end, and step. If start is omitted, the slice starts at the beginning. If end is omitted, the slice goes to the end. If step is omitted, it is one.

```python
# some list slicing examples
fruits = ['rambutan', 'jackfruit', 'lychee', 'apple', 'orange']
print(fruits[1:3]) # second and third elements (indeces 1 and 2)
print(fruits[2:]) # from third to end
print(fruits[:3]) # first three (indexes 0, 1, 2)
# the following is another way to get the first three:
print(fruits[-5:-2]) # counting from the end; index -1 refers to the last item
print(fruits[-4:3]) # you can mix negative and positive indeces
print(fruits[-4:0]) # you get the empty list if the second index is before the first
print(fruits[3:2]) # again the empty list
```

## Slicing Strings

Slicing strings gives you substrings, like slicing lists gives you sublists.

```python
s="is this a fruit?"
print(s[-6:]) # just "fruit?"
print(s[:2]) # just "is"
print(s[3:7]) # just "this"
print(s[30:])
```

## Chaining Slices and Indexes

You can first slice, then index, or vice versa as in the following example:

```python
fruits = ['rambutan', 'jackfruit', 'lychee', 'apple', 'orange']
print(fruits[-3:]) # we get a list back
print(fruits[-3:][-1]) # this returns the last string ("orange") in that list
print(fruits[1:]) # another sublist
print(fruits[1:][3]) # another to way to get "orange"
print(fruits[1:][-1][:2]) # first two letters in "orange"
print(fruits[1:][3][:-4]) # another way to get the first two letters in "orange"
```

## Python List Methods

Note the lists are mutable (changeable) so that these methods modify a list in place.

- append: add an item to the end of the list
- extend: add a list of items to the end of the list
- insert: insert an item at a specified position
- remove: remove the first instance of a specified item from a list
- pop: remove an item at a given position in the list and return it; if position is not specified, remove and return the last item
- clear: remove everything from the list
- index: return the index of the first occurrence of a specified item
- count: count how many there are of a particular item
- sort: sort the items in the list
- reverse: reverse the list
- copy: make a shallow copy of the list

## Some Examples of List Methods

Here is some code that uses some of the list methods.

```python
nums=[5,6,7,8,3,9]
nums.remove(7)
print(nums) # now [5, 6, 8, 3, 9]
print(nums.pop()) # prints and removes 9
print(nums) # now [5, 6, 8, 3]
nums.append(6)
print(nums) # now [5, 6, 8, 3, 6]
print(nums.index(6)) # returns 1 which is the index of the first 6
nlist=[2,2,3,3,3,4,4,4,5,5]
print(nlist.count(4)) # 3
print(nlist.count(5)) # 2
nlist.remove(3)
print(nlist.count(3)) # 2
try:
    nlist.remove(1)
except ValueError as e:
    print(type(e), e)
```

## Sorting, Maximum, Minimum

The sort method sorts in place.

The sorted function creates and returns a new object.

Min and max give the minimum and maximum items in a list.

```python
l=[5,4,3,2,1]
l.sort()
print(l) # l is changed

l2=[5,4,3,2,1]
l3=sorted(l2)
print(l2) # l2 is unchanged
print(l3) # l3 is new
print(min(l3)) # 1
print(max(l3)) # 5
```

## The "Any" and "All" Functions

The "any" and "all" functions see whether any or all of the objects in a list are true, based on the truth value of the objects. From the Python manual, here are most of the built-in objects considered false:

- constants defined to be false: None and False
- zero of any numeric type: 0, 0.0, 0j, Decimal(0), Fraction(0,1)
- empty sequences and collections: '',(),[],{},set(), range(0)

Most other objects are considered True.

```python
l=[5,4,True,100,1,[0]]
print(all(l)) # True; note [0] is not False because it is not an empty list
l.append(0)
print(all(l)) # now False
print(any(l)) # True
l2=[False,0,0,False,[],{}]
print(any(l2)) # False
l2.append(5)
print(any(l2)) # True
print(all(l2)) # False
```

## Lists versus Strings

What are some differences between lists and strings?

- strings are immutable: strings cannot be changed
- string methods return a new value, while list methods often change the original list (and don't return a value)
- (and obviously, they are different types of objects that contain different things)

```python
greeting="hello"
try:
    greeting[0]="m"
except:
    print("can't change a string")

g2=greeting.upper() # can create a new string using a string method
print(greeting) # "hello"
print(g2) # "HELLO"
g3='m'+greeting[1:] # creates a new string
print(g3) # "mello"
print(greeting) # "hello"
```

## Converting Strings to Lists and Back Again

You can split a string on a delimiter and store the parts in a list. Here's an example:

```python
# split string to list
created_date = "09/02/2020"
created_date_parts = created_date.split('/')
print(created_date_parts)
print(type(created_date_parts))

# extract individual strings
month, day, year = created_date_parts
print('month and year are:', month, year)
print(type(month)) # str
print(type(year)) # str

# reassemble (in a different order, using join and specifying the delimiter)
ymd='-'.join([year, month, day])
print(ymd)
print(type(ymd)) # str
```

## Range Objects

A range is an immutable sequence object.

- You can loop over it
- You can index into it
- You can't assign to it, because it is immutable

The elements are not stored in memory but are generated as needed. Therefore you can have a very large range object

```python
r = range(5)
print(r)
print(list(r)) # convert it to list
print(type(r))
print(r[2])
#r[2]=5
r2=range(0,10000000000,5) # you can create a huge range 
print(r2[100000001])
print(type(r2))
#l=list(r2) # trying to do this will eat up all of your memory and can crash your machine.
for i in r: # can loop through a range
    print(i, end=' ')

# trying to assign to it won't work
try:
    r[0] = 21
except TypeError as e:
    print(type(e), e)

# to assign, first have to convert it to a list
l_from_r=list(r)
print(l_from_r)
l_from_r[2]=100
print(l_from_r)
```

## Tuples

Tuples are similar to lists, but unlike lists are immutable.

They take up less memory than lists and perform better. They are good for data that won't change.

Tuples can be created with commas alone.

```python
# only commas
t = 1, 2, 3

# with one element, element and comma
one_element = 2,

print(t)
print(type(t))
print(one_element)
print(type(one_element))
```

## Sometimes You Need Parentheses ...

For instance, if printing a tuple literal:

```python
# if you want a tuple literal as an argument
# ...it has to be wrapped in parentheses
print((1, 2, 3), 4)
```

## Tuples are Immutable

```python
t= "apple", "orange"
try:
    t[0] = 'pear'
except TypeError as e:
    print(e)
```

## Tuples: Useful for Assignments

```python
word1, word2 = 'foo', 'bar'
print(word1)
print(word2)

x, y = 1, 2
x, y = y, x
print(x)
print(y)
```

## More Tuple Assignment Examples

```python
x=(["a","b"],["c","d"]) # tuple of two lists
print(x)
print(type(x))
print(type(x[0]))

t2 = 'foo', 'bar'
word1, word2 = t2
print(word1)
print(word2)
```

## A List of Tuples

```python
points = [(1, 2), (3, 4), (5, 6)]
for p in points:
    # type is tuple on each iteration...
    print(type(p))
    print(p)

# unpacking in the loop variable
for x, y in points:
    print('x', x)
    print('y', y)
```

## A tuple does not equal an equivalent list

```python
print([1,2,3]==(1,2,3))
print(tuple([1,2,3])==(1,2,3))
```

## What If You Want the Index and the Element?

```python
# using range
options = ['yes', 'no', 'maybe']
for i in range(len(options)):
    print(i, options[i])
```

You could use range as above, but enumerate does this more compactly:

```python
options = ['yes', 'no', 'maybe']
enum=enumerate(options)
print(enum) # enumerate object
print(list(enum)) # converted to list of tuples
for i, option_label in enumerate(options):
    print(i, option_label)
```

## Dictionaries

In some other languages called hashes or associative arrays

- type and constructor is dict
- a dictionary is a collection of pairs of keys and their associated values
- empty dictionary is {}
- values can be looked up with keys into using [] as in d[k]
- new key / val pair can be added with [] and assignment, for example d[new\_key]=new\_value
- KeyError if key doesn't exist
- built using hash tables: close to constant time assignment and lookup, so typically very fast

```python
capitals={"New York":"Albany","California":"Sacramento"}
print(capitals["New York"])
```

## An Example of a Dictionary

```python
birth_years =  dict()
birth_years['Beethoven'] = 1770
birth_years['Bach']=1685
print(birth_years['Beethoven'])
print(birth_years)

try:
    print(birth_years['Mozart'])
except KeyError:
    print('error: that key does not exist')
```

## Dictionaries and Printing

Let's print the following dictionary's keys and values:

```python
person = {"first":"james", "last":"jones", "middle": "earl"}
for name_part in person:
    print(name_part) # print key
for name_part in person:
    print(person[name_part]) # print value
```

## Using .items

```python
person = {"first":"james", "last":"jones", "middle": "earl"}
print(list(person.items()))
print(type(person.items())) # like a list of key, value tuples
for k, v in person.items():
    print(k,v)
```

## .values and .keys

```python
person = {"first":"james", "last":"jones", "middle": "earl"}
vals=list(person.values())
print(vals)
keys=list(person.keys())
print(keys)
```

## No Duplicate Keys

```python
d_with_dup={'John':20, 'Vinod':34, 'John':17}
print(d_with_dup) # first value is overwritten
```

## Sets

Sets are an unordered collection of distinct elements.

```python
# use curly braces
words = {'apple', 'peach', 'banana', 'apple'}
print(words)
print(type(words))

# an empty set
empty=set()
print(type(empty))

# create a set of the numbers from 1 to 10
numbers=set(range(1,11))
print(numbers) # note braces in output
print(list(numbers)) # square brackets
print(tuple(numbers)) # parentheses
```

## Set Operators and Methods: Union and Intersection

```python
words1={"apple", "peach", "banana"}
words2={"orange","pear"}
words3={"banana","pineapple"}

print(words1.union(words2))
print(words2.union(words1))
print(words1 | words2)

print(words1.intersection(words3))
print(words3.intersection(words1))
print(words1 & words3)
```

## Sets: Immutable, Distinct Elements

```python
try:
    # set elements cannot be mutable!
    {'foo', [1, 2, 3]}
except TypeError as e:
    print(type(e), e) # sets, like dictionaries, are implemented with hashes

print(set([1,2,2,2,3])) # each element only goes in once
```

## Sets: Interaction with Other Types

```python
words= {"apple", "orange", "banana", "peach"}
# can use union and intersection with other iterable types like lists or tuples

print(words.union([1, 2, 3, 3]))
print(words.union((5,6,6,7,7,7)))

try:
    # set operators will raise exception if the operands are of different types
    words | [1, 2, 2]
except TypeError as e:
    print(type(e), e)

# fix by casting
print(words | set([1,2,2]))
```

## Lists and Tuples are Ordered; Sets are Not

```python
# ordered
print([1,2,3]==[3,2,1]) # False
print((1,2,3)==(3,2,1)) # False
# unordered
print({1,2,3}=={3,2,1}) # True

# sets are not equal to tuples or lists
print({1,2,3}==(1,2,3)) # False: set != tuple
print({1,2,3}==set((1,2,3))) # True
```
