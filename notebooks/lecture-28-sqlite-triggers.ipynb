This defines a trigger which updates the customer address in the orders table when it is updated in the customers table. Note that "old.name" and "new.address" refer to the old value of the name in the customers table and the new value of the address in the customers table, respectively.

```python
import sqlite3
import pandas as pd
import os
try:
  os.remove("customers_orders.db") # remove the database if it is already there
  print("removed earlier copy of customers_orders.db")
except:
  print("database not already there")
# create a connection to a database
conn = sqlite3.connect("customers_orders.db") # creates the database if it is not already there.
def execSQL(conn,query):
  conn.execute(query) # execute an SQL query
  conn.commit() # "commit" that query in order to make its action permanent
def allrowsSelect(conn,query):
  cursor = conn.execute(query)
  for row in cursor:
   print(row)
execSQL(conn,"PRAGMA foreign_keys=ON")


# put the SQL to create the tables in a string
createCMD='''
CREATE TABLE if not exists customers(
  customerid    INTEGER PRIMARY KEY,
  name TEXT,
  address TEXT
)
'''
# execute it
execSQL(conn,createCMD)
query='''
INSERT INTO customers (name, address)
	VALUES
	  ('Joe Klein','2 Park Avenue'),
	  ('Mary Liu', '4 Fifth Avenue')
'''
execSQL(conn,query)

# put the SQL to create the tables in a string
createCMD='''
CREATE TABLE if not exists orders(
  orderid    INTEGER PRIMARY KEY,
  description  TEXT,
  customer_name TEXT,
  customer_address TEXT,
  customerid INTEGER,
  FOREIGN KEY(customerid) REFERENCES customers(customerid)
)
'''
# execute it
execSQL(conn,createCMD)
query='''
INSERT INTO orders (description,customer_name,customer_address,customerid)
	VALUES
	  ('Thesaurus', 'Mary Liu','4 Fifth Avenue',2),
	  ('Dictionary','Joe Klein','2 Park Avenue',1)
'''
execSQL(conn,query)
s="""
CREATE TRIGGER update_customer_address UPDATE OF address ON customers
  BEGIN
    UPDATE orders SET customer_address = new.address WHERE customer_name = old.name;
  END;
"""
s=conn.executescript(s)
print(s)
```

```python
import sqlite3
import pandas as pd
import os
try:
  os.remove("customers_orders.db") # remove the database if it is already there
  print("removed earlier copy of customers_orders.db")
except:
  print("database not already there")
# create a connection to a database
conn = sqlite3.connect("customers_orders.db") # creates the database if it is not already there.
def execSQL(conn,query):
  conn.execute(query) # execute an SQL query
  conn.commit() # "commit" that query in order to make its action permanent
def allrowsSelect(conn,query):
  cursor = conn.execute(query)
  for row in cursor:
   print(row)
execSQL(conn,"PRAGMA foreign_keys=ON")


# put the SQL to create the tables in a string
createCMD='''
CREATE TABLE if not exists customers(
  customerid    INTEGER PRIMARY KEY,
  name TEXT,
  address TEXT
)
'''
# execute it
execSQL(conn,createCMD)
query='''
INSERT INTO customers (name, address)
	VALUES
	  ('Joe Klein','2 Park Avenue'),
	  ('Mary Liu', '4 Fifth Avenue')
'''
execSQL(conn,query)

# put the SQL to create the tables in a string
createCMD='''
CREATE TABLE if not exists orders(
  orderid    INTEGER PRIMARY KEY,
  description  TEXT,
  customer_name TEXT,
  customer_address TEXT,
  customerid INTEGER,
  FOREIGN KEY(customerid) REFERENCES customers(customerid)
)
'''
# execute it
execSQL(conn,createCMD)
query='''
INSERT INTO orders (description,customer_name,customer_address,customerid)
	VALUES
	  ('Thesaurus', 'Mary Liu','4 Fifth Avenue',2),
	  ('Dictionary','Joe Klein','2 Park Avenue',1)
'''
execSQL(conn,query)
s="""
CREATE TRIGGER update_customer_address UPDATE OF address ON customers
  BEGIN
    UPDATE orders SET customer_address = new.address WHERE customer_name = old.name;
  END;
"""
s=conn.executescript(s)
df=pd.read_sql_query("select * from customers",conn,index_col='customerid')
print(df)
df2=pd.read_sql_query("select * from orders",conn,index_col='orderid')
print(df2)
execSQL(conn,"UPDATE customers SET address = '100 Seventh Avenue' WHERE name = 'Joe Klein'")
df=pd.read_sql_query("select * from customers",conn,index_col='customerid')
print(df)
df2=pd.read_sql_query("select * from orders",conn,index_col='orderid')
print(df2)
```
