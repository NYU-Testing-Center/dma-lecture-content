In statistics, the process of filling in missing data, and of dealing with missing data in general, is called "imputation." This is a very complex topic and whole courses and [books](https://stefvanbuuren.name/fimd/) are concerned with the topic.

The simplest imputation approach is to replace each missing value with the column's mean. In the example below, we do this using a dictionary.

Mean imputation only works with numerical data, not categorical data. And it makes little sense for binary categorical data (like gender, for instance.)

```python
# another toy dataframe
import numpy as np
import pandas as pd
df = pd.DataFrame(np.arange(12).reshape((3, 4)),  columns=list('abcd'))
# make two of the values missing
df.loc[1, 'd'] = np.nan
df.loc[2, 'c'] = np.nan

print("Before Mean Imputation:")
print(df)
# sets missing values in each column to the mean of that column
# mean computation ignores missing values
print("After Mean Imputation")
df.fillna({'c':df['c'].mean(),'d':df['d'].mean()},inplace=True)
print(df)
```

["Hot deck"](https://pubmed.ncbi.nlm.nih.gov/21743766/) imputation just fills missing values with the previous non-missing value in the column. This can be reasonable if you expect similar observations (that is, rows) to be near one another in the list or if you have sorted them so that this is the case.

In pandas, you can use the ffill (forward fill) method to accomplish this. There is also bfill which files with the next non-missing observation rather than the previous one.

```python
# another toy dataframe
import numpy as np
import pandas as pd
df = pd.DataFrame(np.arange(12).reshape((3, 4)),  columns=list('abcd'))
# make two of the values missing
df.loc[1, 'd'] = np.nan
df.loc[2, 'c'] = np.nan

print("Before Hot Deck Imputation:")
print(df)
# sets missing values in each column to the mean of that column
# mean computation ignores missing values
print("After Hot Deck Imputation (forward)")
df2=df.ffill()
print(df2)
print("After Hot Deck Imputation (backward)")
df3=df.bfill()
print(df3)
```

Perhaps you have a lot of continuous data. It may be useful to categorize that data into bins for easier analysis or reporting.

To do this, you use pd.cut(data,bins,labels=labels)

Where the bins are the boundaries of the bins as a list.

labels can be used to specify text labels for each bin.

This creates a categories object which can be counted with the value\_counts method.

In the example below, we generate 100 random numbers between 60 and 100 representing student grades. We then bin them into letter grades where 60-69 is a D, 70-79 is a C, etc. We can then count the categories using value\_counts.

```python
import pandas as pd
import numpy as np
# generate 100 random numbers between 60 and 100 (simulating student grades)
scores=np.random.randint(60,101,size=100)
# represents 60-69 (D), 70-79 (C), 80-89 (B), 90-100 (A)
# upper is inclusive, lower is exclusive
bins=[59,69,79,89,100]
gradelabels=['D','C','B','A' ]
print(scores)
grades=pd.cut(scores,bins, labels=gradelabels)
print(grades)
print(type(grades))
print(grades.value_counts())
```

In the code below, we create a dataframe.

First, we find the rows with values in their second column over 10.

Then we replace all of those values in the second column that are over 10 with 10. This is sometimes called top-coding.

```python
import pandas as pd
import numpy as np
df = pd.DataFrame([[2, 3], [1, 50], [20, 4], [3, 45]])
print("Original dataframe:")
print(df)
print("Rows with values of the Second Column over 10:")
print(df[df[1]>10])
df.loc[df[1]>10,1] = 10
print("Dataframe with replaced values:")
print(df)
```

We read in a dataframe below. Then we remove the dollar signs and convert to numeric, coercing anything that doesn't convert (e.g. a string) to NaN.

```python
import pandas as pd
data = [['2009', '$500'],
        ['2010', '$1,234'],
        ['2011', 'bad data'],
        ['2012', '$2,507']]
df = pd.DataFrame(data , columns=['date', 'total'])
print("Original Dataframe:")
print(df)
df['total'] = df['total'].str.replace('$', '', regex=False)
df['total'] = df['total'].str.replace(',', '',regex=False)
df['total'] = pd.to_numeric(df['total'],errors='coerce')
print("Modified Dataframe:")
print(df)
```

The pandas merge method joins data from one DataFrame with another based on common column values. The default result is the intersection of rows with matching column values, with all columns included, as in the example below. This is called an inner join.

Three ways to do the same inner join are shown.

```python
import pandas as pd
a = pd.DataFrame([[2, 20], [4, 40], [6, 60], [8, 80]],
                 columns=['k', 'col1'])
b = pd.DataFrame([[4, 2], [4, 3], [8, 7], [10,6]],
                 columns=['k', 'col2'])
print("First Dataframe:")
print(a)
print("Second Dataframe:")
print(b)
print("Inner Join: Method 1")
print(pd.merge(a,b,on='k'))
print("Inner Join: Method 2")
print(a.merge(b,on='k'))
print("Inner Join: Method 3")
print(b.merge(a,on='k'))
```

    Types of Merges (Joins)
  </div>

The following are the main types of joins in use:

- 'left': include all keys from the first (left) DataFrame, even if they don't exist in second (right)
- 'right': include all keys from the second (right) DataFrame, even if they don't exist in first (left)
- 'outer': union of keys from both DataFrames
- 'inner': intersection of keys from both DataFrames (default)

We will see all of these again when we study SQL.

```python
import pandas as pd
a = pd.DataFrame([[2, 20], [4, 40], [6, 60], [8, 80]],
                 columns=['k', 'col1'])
b = pd.DataFrame([[4, 2], [4, 3], [8, 7], [10,6]],
                 columns=['k', 'col2'])
print("First Dataframe:")
print(a)
print("Second Dataframe:")
print(b)
print("Outer Join:")
print(pd.merge(a,b,on='k',how='outer'))
print("Left Join:")
print(pd.merge(a,b,on='k',how='left'))
print("Right Join:")
print(pd.merge(a,b,on='k',how='right'))
print("Inner Join:")
print(pd.merge(a,b,on='k',how='inner'))
```

```python
import pandas as pd
import numpy as np
d1 = pd.DataFrame(np.arange(9).reshape((3, 3)),
                columns=list('abc'))
d2 = pd.DataFrame(np.arange(10, 19).reshape((3, 3)),
                columns=list('abc'))
print(d1)
print(d2)
d3=pd.concat([d1,d2])
print(d3)
d3.index=range(6) # resetting the index
print(d3)
```
