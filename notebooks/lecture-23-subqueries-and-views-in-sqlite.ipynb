Let's find the biggest gap between the event date and the day it was added to CAERS (created\_date), using the julianday function in SQLite that converts a string representing a date to a number of days since the January 1, 4713 BC.

To start, let's just look at the date fields.

```python
import sqlite3
import pandas as pd
conn=sqlite3.connect("caers.db")
# first let's just look at the date fields
query="""select julianday(created_date), created_date,
julianday(event_date), event_date, product from caers limit 10"""
print(pd.read_sql_query(query,conn))
```

Now we find the row with the maximum difference between the created data and the event date.

```python
import sqlite3
import pandas as pd
conn=sqlite3.connect("caers.db")
# first let's just look at the date fields
query="""select max(julianday(created_date)-julianday(event_date)) as maxdiff,
event_date, created_date, product from caers"""
print(pd.read_sql_query(query,conn))
```

Let's find how many records have non-null created\_date and how many have non-null event\_date. This involves a sub-query. A sub-query is a query within a query, that is, in this case, a select within a select.

We can see that all records have a created date, but not all have an event date.

Also note that there is no FROM clause in the outer SELECT statement.

```python
import sqlite3
import pandas as pd
conn=sqlite3.connect("caers.db")
query='''
select
       (select count(*) from caers) as record_count,
       (select count(*) from caers where created_date is not null) as created_count,
       (select count(*) from caers where event_date is not null) as event_count
'''
print(pd.read_sql_query(query,conn))
```

Let's find the number of events by year, and then limit it to the top five. Here we are using the SQLite strftime function to extract the year from the date string.

```python
import sqlite3
import pandas as pd
conn=sqlite3.connect("caers.db")
query='''
select strftime('%Y', event_date) as year, count(*) as c
from caers group by year
having year is not null
order by c desc limit 5;
'''
print(pd.read_sql_query(query,conn))
```

Let's find the number of events in the year with the most events. This involves a sub-query, that is, a select from the results of a select.

```python
import sqlite3
import pandas as pd
conn=sqlite3.connect("caers.db")
query='''
select year, max(c) from
(select strftime('%Y', event_date) as year, count(*) as c
from caers
group by year
having year is not null)
'''
print(pd.read_sql_query(query,conn))
```

Let's define a view (a "virtual table") holding the counts of events by year. Note views are permanent but do not take up additional space (except for the small amount of space required by the code that defines them), and they can used in the same way as actual tables.

First we delete the view if it is already there. Then we create it. Then we query it to redo the two query results from the two previous slides.

```python
# This example works on its own scratch copy of the data, so it runs the
# same way no matter which of the other examples you ran first.
import shutil
shutil.copyfile("caers.db", "caers-scratch.db")
import sqlite3
import pandas as pd
conn=sqlite3.connect("caers-scratch.db")
def execSQL(conn,query):
  conn.execute(query) # execute an SQL query
  conn.commit()
query='''
drop view if exists year_counts
'''
execSQL(conn,query)
query='''
create view year_counts as
select strftime('%Y', event_date) as year, count(*) as c
from caers group by year
having year is not null
order by c desc
'''
execSQL(conn,query)
print(pd.read_sql_query("select * from year_counts limit 5", conn))
print(pd.read_sql_query("select year, max(c) from year_counts", conn))
```

Find the first five events occuring in the year with the maximum number of events. This involves a sub-query.

```python
# This example works on its own scratch copy of the data, so it runs the
# same way no matter which of the other examples you ran first.
import shutil
shutil.copyfile("caers.db", "caers-scratch.db")
import sqlite3
import pandas as pd
conn=sqlite3.connect("caers-scratch.db")
def execSQL(conn,query):
  conn.execute(query) # execute an SQL query
  conn.commit()
query='''
drop view if exists year_counts
'''
execSQL(conn,query)
query='''
create view year_counts as
select strftime('%Y', event_date) as year, count(*) as c
from caers group by year
having year is not null
order by c desc
'''
execSQL(conn,query)
#Let's query the view.
query='''
select report_id, created_date, product, description from caers
where strftime('%Y', event_date) = (
    select year from year_counts where c =(select max(c) from year_counts)
) limit 5
'''
print(pd.read_sql_query(query,conn))
```
