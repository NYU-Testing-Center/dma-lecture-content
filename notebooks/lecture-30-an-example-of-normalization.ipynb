## Normalizing the Data

The student enrollments table is unnormalized and has lots of duplicate data. How can we normalize it to remove all duplicate data? We can break it into five tables:

- Student information
- Course information
- Professor information
- Student-Course enrollments (keys only)
- Courses each Professor teaches (keys only)

```python
# This example works on its own scratch copy of the data, so it runs the
# same way no matter which of the other examples you ran first.
import shutil
shutil.copyfile("sample_enrollments.db", "sample_enrollments-scratch.db")
import sqlite3
import pandas as pd
def execSQL(conn,query):
  conn.execute(query) # execute an SQL query
  conn.commit() # "commit" that query in order to make its action permanen
# create a connection to a database
conn = sqlite3.connect("sample_enrollments-scratch.db")
q='''
create table courses as
select distinct course_number, course_title, loc, meeting_time
    from sample_enrollments
'''
execSQL(conn,q)
q='''
create table professors as
select distinct professor_name, professor_office
    from sample_enrollments
'''
execSQL(conn,q)
execSQL(conn,"alter table professors rename to prof_temp")
q='''
create table professors(
  prof_id integer primary key,
  professor_name text,
  professor_office text
)
'''
execSQL(conn,q)
execSQL(conn,"insert into professors (professor_name, professor_office) select * from prof_temp")
execSQL(conn,"drop table prof_temp")
print(pd.read_sql_query("select * from professors",conn))
```
